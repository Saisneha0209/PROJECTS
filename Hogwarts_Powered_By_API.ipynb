```python
%%html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>Hogwarts Explorer Pro</title>
    <style>
        body {
            background-color: #1a1a2e;
            color: #e2e2e2;
            font-family: sans-serif;
            text-align: center;
            padding: 20px;
        }
        .controls {
            margin-bottom: 25px;
        }
        #search-bar {
            padding: 12px;
            width: 320px;
            border-radius: 25px;
            border: 2px solid #1f4068;
            background-color: #162447;
            color: white;
            font-size: 16px;
            margin-bottom: 15px;
            outline: none;
            transition: 0.3s;
        }
        #search-bar:focus {
            border-color: gold;
            box-shadow: 0 0 8px gold;
        }
        .house-filters {
            display: flex;
            justify-content: center;
            gap: 10px;
            flex-wrap: wrap;
        }
        .btn {
            background-color: #162447;
            color: #fff;
            border: 2px solid #1f4068;
            padding: 8px 16px;
            cursor: pointer;
            border-radius: 20px;
            font-weight: bold;
            transition: 0.2s;
        }
        /* House specific button highlights */
        .btn[data-house="Gryffindor"]:hover { background-color: #7f0909; border-color: #ffc500; color: #ffc500; }
        .btn[data-house="Slytherin"]:hover { background-color: #0d6217; border-color: #aaaaaa; color: #aaaaaa; }
        .btn[data-house="Hufflepuff"]:hover { background-color: #eee117; border-color: #000000; color: #000000; }
        .btn[data-house="Ravenclaw"]:hover { background-color: #000a90; border-color: #946b2d; color: #946b2d; }
        .btn[data-house="all"]:hover { background-color: gold; color: #1a1a2e; }

        #characters-container {
            display: grid;
            grid-template-columns: repeat(auto-fill, minmax(200px, 1fr));
            gap: 20px;
            padding: 20px;
        }
        .card {
            background-color: #1f4068;
            border-radius: 8px;
            padding: 15px;
            transition: transform 0.2s;
        }
        .card:hover {
            transform: scale(1.05);
        }
        .card img {
            width: 100%;
            height: 250px;
            object-fit: cover;
            border-radius: 4px;
        }
    </style>
</head>
<body>

    <h1>Wizarding World Explorer</h1>

    <div class="controls">
        <input type="text" id="search-bar" placeholder="Search characters by name..."><br>
        <div class="house-filters">
            <button class="btn" data-house="all">Show All</button>
            <button class="btn" data-house="Gryffindor">Gryffindor</button>
            <button class="btn" data-house="Slytherin">Slytherin</button>
            <button class="btn" data-house="Hufflepuff">Hufflepuff</button>
            <button class="btn" data-house="Ravenclaw">Ravenclaw</button>
        </div>
    </div>

    <main id="characters-container">
        <p>Loading wizarding world...</p>
    </main>

    <script>
        const container = document.getElementById('characters-container');
        const searchBar = document.getElementById('search-bar');
        const filterButtons = document.querySelectorAll('.btn');

        let allCharacters = [];
        let activeHouseFilter = 'all';

        async function fetchCharacters() {
            try {
                const response = await fetch('https://hp-api.onrender.com/api/characters');
                const data = await response.json();

                // Only keep characters that have an image
                allCharacters = data.filter(character => character.image !== "");

                displayCharacters(allCharacters);
            } catch (error) {
                container.innerHTML = `<p>Error loading data.</p>`;
                console.error(error);
            }
        }

        function displayCharacters(characters) {
            container.innerHTML = "";

            if (characters.length === 0) {
                container.innerHTML = `<p>No wizards found matching those criteria!</p>`;
                return;
            }

            characters.forEach(character => {
                const card = document.createElement('div');
                card.classList.add('card');

                card.innerHTML = `
                    <img src="${character.image}" alt="${character.name}">
                    <h3>${character.name}</h3>
                    <p>House: ${character.house || 'Unknown'}</p>
                `;
                container.appendChild(card);
            });
        }

        // Combined Filter Function (Handles both Search AND House clicks together!)
        function filterData() {
            const searchString = searchBar.value.toLowerCase();

            const filtered = allCharacters.filter(character => {
                const matchesSearch = character.name.toLowerCase().includes(searchString);
                const matchesHouse = activeHouseFilter === 'all' || character.house === activeHouseFilter;

                return matchesSearch && matchesHouse;
            });

            displayCharacters(filtered);
        }

        // Listen for typing in the search bar
        searchBar.addEventListener('input', filterData);

        // Listen for clicks on the house buttons
        filterButtons.forEach(button => {
            button.addEventListener('click', (e) => {
                activeHouseFilter = e.target.getAttribute('data-house');
                filterData();
            });
        });

        fetchCharacters();
    </script>
</body>
</html>
```
