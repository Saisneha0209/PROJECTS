```python


import os
from getpass import getpass
from copy import deepcopy
from groq import Groq


os.environ["GROQ_API_KEY"] = getpass(
    "Enter your Groq API key: "
)

client = Groq(
    api_key=os.environ["GROQ_API_KEY"]
)


MODEL_NAME = "openai/gpt-oss-120b"


current_concept = "Magnetic poles"

learning_objective = (
    "Understand that like poles repel "
    "and unlike poles attract."
)

def create_student_model():

    return {

        "concept": current_concept,

        "student_state": "NOT_STARTED",

        "support_level": 1,

        "phase": "DIAGNOSIS",

        "misconception": None,

        "misconception_status": None,

        "evidence": [],

        "evidence_matrix": {

            "prior_knowledge": "PENDING",

            "prediction": "PENDING",

            "explanation": "PENDING",

            "guided_application": "PENDING",

            "independent_retest": "PENDING",

            "transfer": "PENDING"
        },

        "difficulty": "Medium",

        "next_action": "PROBE",

        "last_classification": None,

        "success_count": 0,

        "struggle_count": 0,

        "retest_passed": False,

        "transfer_passed": False,

        "advance_allowed": False,

        "decision_reason": "",

        "expected_next_evidence": (
            "Reveal prior knowledge or an initial prediction."
        )
    }


student_model = create_student_model()

print("\n" + "=" * 70)
print("             CLASS 6 SCIENCE AI TUTOR")
print("=" * 70)

print("""
Hello!!

Welcome to your Class 6 Science AI Tutor.

I will connect Science with things around you,
help you observe carefully,
make predictions,
test ideas safely,
and gradually discover deeper explanations.

Science is not only about finding answers.
It is also about learning how to ask
better questions.

Type 'exit' to stop.
""")

def classify_student_response(
    current_concept,
    learning_objective,
    tutor_question,
    student_response,
    history_string,
    expected_next_evidence
):

    classifier_prompt = f"""

You are the Student Response Classifier
inside a Class 6 Science AI Tutor.

Your ONLY job is to determine
what the student's response actually proves.

Do NOT tutor.

Do NOT give hints.

Do NOT explain the lesson.

Do NOT decide mastery.

Do NOT invent evidence.


CURRENT CONCEPT:

{current_concept}


LEARNING OBJECTIVE:

{learning_objective}


TUTOR'S PREVIOUS QUESTION:

{tutor_question}


EXPECTED EVIDENCE:

{expected_next_evidence}


STUDENT RESPONSE:

{student_response}


RECENT CONVERSATION:

{history_string}


================================================
BEFORE CLASSIFYING
================================================

Silently determine:

1. What exactly did the Tutor ask?

2. What evidence was requested?

3. What exactly did the student answer?

4. Did the student actually provide
   the requested evidence?

5. Is the conclusion scientifically correct?

6. Is the reasoning scientifically correct?

7. Does the response reveal:

   a misconception,

   vocabulary problem,

   prerequisite gap,

   uncertainty,

   incomplete reasoning,

   or insufficient evidence?


================================================
ALLOWED CLASSIFICATIONS
================================================

Choose ONLY ONE:

CORRECT

PARTIALLY_CORRECT

INSUFFICIENT_EVIDENCE

INCORRECT

MISCONCEPTION

VOCABULARY_ISSUE

PREREQUISITE_GAP

UNCERTAIN

IRRELEVANT


================================================
CORRECT
================================================

Use CORRECT only when the student
correctly answers the actual question
AND provides the important evidence requested.

A scientifically correct sentence
is NOT automatically a correct answer.

If reasoning was requested,
reasoning must also be correct.


================================================
PARTIALLY_CORRECT
================================================

Use when the student demonstrates
part of the required understanding
but something important is missing
or contains a smaller conceptual error.


Example:

"Like poles repel because they are different."

The conclusion is correct.

But "because they are different"
shows incorrect reasoning.

Do NOT classify this as fully CORRECT.


================================================
INSUFFICIENT_EVIDENCE
================================================

Use when the response might contain
something scientifically correct
but does not provide enough evidence
to answer what was requested.


Example:

Tutor:

"An unknown pole attracts a north pole.
What is the unknown pole and why?"


Student:

"They attract."


The student did not identify the pole
and did not explain why.

Therefore:

INSUFFICIENT_EVIDENCE


================================================
INCORRECT
================================================

Use when the answer is scientifically incorrect
but there is not enough evidence
to identify a stable incorrect mental model.


================================================
MISCONCEPTION
================================================

Use when the response demonstrates
a specific incorrect mental model
or incorrect general rule.


Examples:

"All magnets attract each other."

"North is positive and south is negative."

"North and north attract because both are magnets."

"Field lines physically push magnets apart."


IMPORTANT:

Magnetic north and south poles
are NOT positive and negative electrical charges.

Magnetic field lines are representations
used to describe magnetic fields.

They are NOT physical strings
that push magnets.


================================================
VOCABULARY_ISSUE
================================================

Use when the main difficulty
is understanding a scientific word.

Example:

"What does repel mean?"


================================================
PREREQUISITE_GAP
================================================

Use when the student lacks earlier knowledge
needed for the current objective.


Examples:

"I don't know what a magnetic pole is."

"What is a pole?"

"I don't know anything about this.
Teach me from the basics."

"Can you explain what a magnet is first?"


IMPORTANT:

A specific request for missing foundations
should NOT merely be classified UNCERTAIN.


================================================
UNCERTAIN
================================================

Use when the student is uncertain
without revealing a specific prerequisite gap.


Examples:

"I don't know."

"I'm not sure."

"Maybe."


================================================
IRRELEVANT
================================================

Use when the response does not meaningfully
address the current learning interaction.


================================================
CRITICAL VALIDATION RULES
================================================

A correct sentence
is not automatically a correct answer.

A correct answer
is not automatically mastery.

A lucky guess
is not mastery.

Correctness after a hint
is not independent evidence.

Missing requested reasoning
means the evidence is incomplete.

Do not infer a misconception
when there is no evidence for one.

Do not infer mastery
from one correct response.


================================================
OUTPUT FORMAT
================================================

Return EXACTLY:

CLASSIFICATION: one_allowed_classification
REASON: one short sentence

"""

    response = client.chat.completions.create(

        messages=[
            {
                "role": "user",
                "content": classifier_prompt
            }
        ],

        model=MODEL_NAME
    )

    return response.choices[0].message.content


# ================================================================
# 6. PARSE CLASSIFIER
# ================================================================

def parse_classification(classification_result):

    classification = "UNKNOWN"

    reason = ""

    for line in classification_result.splitlines():

        line = line.strip()

        if line.startswith("CLASSIFICATION:"):

            classification = (
                line.replace(
                    "CLASSIFICATION:",
                    ""
                )
                .strip()
                .upper()
            )

        elif line.startswith("REASON:"):

            reason = (
                line.replace(
                    "REASON:",
                    ""
                )
                .strip()
            )

    return classification, reason


# ================================================================
# 7. SUPPORT FUNCTIONS
# ================================================================

def increase_support(student_model):

    if student_model["support_level"] < 6:

        student_model["support_level"] += 1

    return student_model


def fade_support(student_model):

    if student_model["support_level"] > 1:

        student_model["support_level"] -= 1

    return student_model


# ================================================================
# 8. ADVANCEMENT GATE
# ================================================================

def can_advance(student_model):

    evidence = student_model["evidence_matrix"]

    required_evidence = [

        evidence["explanation"] == "PASSED",

        evidence["independent_retest"] == "PASSED",

        evidence["transfer"] == "PASSED"
    ]

    return all(required_evidence)


# ================================================================
# 9. EVIDENCE MATRIX UPDATE
# ================================================================

def update_evidence_matrix(
    student_model,
    classification,
    previous_action,
    previous_phase,
    previous_expected_evidence
):

    evidence = student_model["evidence_matrix"]

    expected = (
        previous_expected_evidence
        or ""
    ).lower()


    # ------------------------------------------------------------
    # PRIOR KNOWLEDGE
    # ------------------------------------------------------------

    # Only meaningful conceptual content counts.
    #
    # "I don't know" does NOT automatically prove
    # prior knowledge.

    if classification in [

        "CORRECT",

        "PARTIALLY_CORRECT",

        "INSUFFICIENT_EVIDENCE",

        "INCORRECT",

        "MISCONCEPTION"

    ]:

        evidence["prior_knowledge"] = "COLLECTED"


    prediction_requested = any(

        phrase in expected

        for phrase in [

            "prediction",

            "predict",

            "what will happen",

            "what happens"

        ]
    )


    if prediction_requested:

        if classification == "CORRECT":

            evidence["prediction"] = "PASSED"

        elif classification in [

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "UNCERTAIN"

        ]:

            evidence["prediction"] = "PARTIAL"

    explanation_requested = any(

        phrase in expected

        for phrase in [

            "explain",

            "reason",

            "why",

            "reasoning",

            "scientifically correct"

        ]
    )


    if explanation_requested:

        if classification == "CORRECT":

            evidence["explanation"] = "PASSED"

        elif classification in [

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE"

        ]:

            evidence["explanation"] = "PARTIAL"


    # ------------------------------------------------------------
    # GUIDED APPLICATION
    # ------------------------------------------------------------

    supported_actions = [

        "NARROW",

        "CONTRAST",

        "HINT",

        "EXAMPLE",

        "EXPLAIN",

        "REMEDIATE_PREREQUISITE"

    ]


    if previous_action in supported_actions:

        if classification == "CORRECT":

            evidence["guided_application"] = "PASSED"


    # ------------------------------------------------------------
    # INDEPENDENT RETEST
    # ------------------------------------------------------------

    if previous_phase == "RETEST":

        if classification == "CORRECT":

            evidence["independent_retest"] = "PASSED"

        elif classification in [

            "INCORRECT",

            "MISCONCEPTION",

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "UNCERTAIN"

        ]:

            evidence["independent_retest"] = "FAILED"


    # ------------------------------------------------------------
    # TRANSFER
    # ------------------------------------------------------------

    if previous_phase == "TRANSFER":

        if classification == "CORRECT":

            evidence["transfer"] = "PASSED"

        elif classification in [

            "INCORRECT",

            "MISCONCEPTION",

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "UNCERTAIN"

        ]:

            evidence["transfer"] = "FAILED"


    return student_model


# ================================================================
# 10. MISCONCEPTION HELPERS
# ================================================================

def activate_misconception(
    student_model,
    reason
):

    student_model["misconception"] = reason

    student_model["misconception_status"] = "ACTIVE"

    return student_model


def reduce_misconception(student_model):

    if (
        student_model["misconception_status"]
        == "ACTIVE"
    ):

        student_model[
            "misconception_status"
        ] = "REDUCED"

    return student_model


def resolve_misconception_if_ready(
    student_model
):

    evidence = student_model["evidence_matrix"]

    if (

        student_model["misconception_status"]
        == "REDUCED"

        and evidence[
            "independent_retest"
        ] == "PASSED"

        and evidence[
            "transfer"
        ] == "PASSED"

    ):

        student_model[
            "misconception_status"
        ] = "RESOLVED"

    return student_model


# ================================================================
# 11. DECISION ENGINE
# ================================================================

def update_tutor_state(
    classification_result,
    student_model
):

    classification, reason = (
        parse_classification(
            classification_result
        )
    )


    student_model[
        "last_classification"
    ] = classification


    previous_action = (
        student_model["next_action"]
    )

    previous_phase = (
        student_model["phase"]
    )

    previous_expected_evidence = (
        student_model[
            "expected_next_evidence"
        ]
    )


    # ============================================================
    # INITIAL QUERY
    # ============================================================

    if classification == "INITIAL_QUERY":

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "support_level"
        ] = 1

        student_model[
            "phase"
        ] = "DIAGNOSIS"

        student_model[
            "next_action"
        ] = "PROBE"

        student_model[
            "decision_reason"
        ] = (
            "This is the beginning of the learning "
            "interaction, so first reveal what "
            "the student already knows."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student reveals prior knowledge, "
            "a prediction, or an initial mental model."
        )

        student_model[
            "advance_allowed"
        ] = False

        return student_model


    # ============================================================
    # UPDATE EVIDENCE
    # ============================================================

    student_model = update_evidence_matrix(

        student_model,

        classification,

        previous_action,

        previous_phase,

        previous_expected_evidence
    )


    evidence_text = (
        f"{classification}: {reason}"
    )


    student_model[
        "evidence"
    ].append(
        evidence_text
    )


    if len(
        student_model["evidence"]
    ) > 20:

        student_model[
            "evidence"
        ].pop(0)


    # ============================================================
    # CORRECT
    # ============================================================

    if classification == "CORRECT":

        student_model[
            "success_count"
        ] += 1

        student_model[
            "struggle_count"
        ] = 0


        # --------------------------------------------------------
        # CORRECT DURING RETEST
        # --------------------------------------------------------

        if previous_phase == "RETEST":

            student_model[
                "retest_passed"
            ] = True

            student_model[
                "student_state"
            ] = "DEVELOPING"

            student_model[
                "support_level"
            ] = 1

            student_model[
                "phase"
            ] = "TRANSFER"

            student_model[
                "next_action"
            ] = "TRANSFER"

            student_model[
                "decision_reason"
            ] = (
                "The student succeeded during an "
                "independent retest. Now test whether "
                "the same concept transfers to a "
                "changed situation."
            )

            student_model[
                "expected_next_evidence"
            ] = (
                "The student independently solves "
                "a changed-context magnetic-pole "
                "inference problem and explains why."
            )


        # --------------------------------------------------------
        # CORRECT DURING TRANSFER
        # --------------------------------------------------------

        elif previous_phase == "TRANSFER":

            student_model[
                "transfer_passed"
            ] = True

            student_model[
                "support_level"
            ] = 1

            student_model[
                "phase"
            ] = "MASTERY_CHECK"

            student_model = (
                resolve_misconception_if_ready(
                    student_model
                )
            )

            student_model[
                "advance_allowed"
            ] = can_advance(
                student_model
            )


            if student_model[
                "advance_allowed"
            ]:

                student_model[
                    "student_state"
                ] = "SECURE"

                student_model[
                    "next_action"
                ] = "ADVANCE"

                student_model[
                    "decision_reason"
                ] = (
                    "The required explanation, "
                    "independent retest, and transfer "
                    "evidence have been demonstrated."
                )

                student_model[
                    "expected_next_evidence"
                ] = (
                    "No further evidence is required "
                    "for this learning objective."
                )

            else:

                student_model[
                    "student_state"
                ] = "DEVELOPING"

                student_model[
                    "next_action"
                ] = "PROBE"

                student_model[
                    "decision_reason"
                ] = (
                    "Transfer succeeded, but another "
                    "required evidence category is "
                    "still missing."
                )

                student_model[
                    "expected_next_evidence"
                ] = (
                    "The student provides the missing "
                    "explanation evidence."
                )


        # --------------------------------------------------------
        # CORRECT AFTER SUPPORT
        # --------------------------------------------------------

        elif previous_action in [

            "NARROW",

            "CONTRAST",

            "HINT",

            "EXAMPLE",

            "EXPLAIN",

            "REMEDIATE_PREREQUISITE"

        ]:

            student_model[
                "student_state"
            ] = "DEVELOPING"

            student_model = fade_support(
                student_model
            )

            student_model[
                "phase"
            ] = "RETEST"

            student_model[
                "next_action"
            ] = "RETEST"


            # Reduce misconception only if the previous
            # requested evidence actually targeted it.

            target_text = (
                previous_expected_evidence
                or ""
            ).lower()


            misconception_targeted = any(

                phrase in target_text

                for phrase in [

                    "incorrect idea",

                    "correct reasoning",

                    "magnetic poles from electric",

                    "revises",

                    "distinguish"

                ]
            )


            if misconception_targeted:

                student_model = (
                    reduce_misconception(
                        student_model
                    )
                )


            student_model[
                "decision_reason"
            ] = (
                "The student succeeded after receiving "
                "support. Guided success is not mastery, "
                "so support must now be removed."
            )

            student_model[
                "expected_next_evidence"
            ] = (
                "The student independently answers "
                "a fresh magnetic-pole case correctly "
                "without hints or choices."
            )


        # --------------------------------------------------------
        # REPEATED INDEPENDENT CORRECT RESPONSES
        # --------------------------------------------------------

        elif (

            previous_action == "PROBE"

            and student_model[
                "success_count"
            ] >= 2

        ):

            student_model[
                "student_state"
            ] = "DEVELOPING"

            student_model[
                "support_level"
            ] = 1

            student_model[
                "phase"
            ] = "TRANSFER"

            student_model[
                "next_action"
            ] = "TRANSFER"

            student_model[
                "decision_reason"
            ] = (
                "The student has provided repeated "
                "independent correct evidence. "
                "Now test changed-context transfer."
            )

            student_model[
                "expected_next_evidence"
            ] = (
                "The student independently solves "
                "a changed-context inference problem "
                "and explains the reasoning."
            )


        # --------------------------------------------------------
        # FIRST CORRECT ANSWER
        # --------------------------------------------------------

        else:

            student_model[
                "student_state"
            ] = "DEVELOPING"

            student_model[
                "support_level"
            ] = 1

            student_model[
                "phase"
            ] = "DIAGNOSIS"

            student_model[
                "next_action"
            ] = "PROBE"

            student_model[
                "decision_reason"
            ] = (
                "The response is correct, but one "
                "correct response does not prove "
                "secure understanding."
            )

            student_model[
                "expected_next_evidence"
            ] = (
                "The student explains why the "
                "answer is scientifically correct."
            )


    # ============================================================
    # PARTIALLY CORRECT
    # ============================================================

    elif classification == "PARTIALLY_CORRECT":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "struggle_count"
        ] += 1

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "phase"
        ] = "GUIDED_PRACTICE"


        if student_model[
            "support_level"
        ] < 2:

            student_model[
                "support_level"
            ] = 2

            student_model[
                "next_action"
            ] = "NARROW"

        else:

            student_model = (
                increase_support(
                    student_model
                )
            )

            level = student_model[
                "support_level"
            ]


            if level == 3:

                student_model[
                    "next_action"
                ] = "CONTRAST"

            elif level == 4:

                student_model[
                    "next_action"
                ] = "HINT"

            elif level == 5:

                student_model[
                    "next_action"
                ] = "EXAMPLE"

            else:

                student_model[
                    "next_action"
                ] = "EXPLAIN"


        student_model[
            "decision_reason"
        ] = (
            "The student demonstrates part of the "
            "idea, so target only the missing or "
            "incorrect part."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student supplies the missing idea "
            "and correct reasoning."
        )


    # ============================================================
    # INSUFFICIENT EVIDENCE
    # ============================================================

    elif classification == "INSUFFICIENT_EVIDENCE":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "phase"
        ] = "DIAGNOSIS"

        student_model[
            "support_level"
        ] = max(
            student_model[
                "support_level"
            ],
            2
        )

        student_model[
            "next_action"
        ] = "NARROW"

        student_model[
            "decision_reason"
        ] = (
            "The response does not provide the "
            "evidence requested, so progression "
            "must remain blocked."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student provides the missing part "
            "of the answer and requested reasoning."
        )


    # ============================================================
    # INCORRECT
    # ============================================================

    elif classification == "INCORRECT":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "struggle_count"
        ] += 1

        student_model[
            "student_state"
        ] = "NEEDS_REVIEW"

        student_model[
            "phase"
        ] = "DIAGNOSIS"


        struggle = student_model[
            "struggle_count"
        ]


        if struggle == 1:

            student_model[
                "support_level"
            ] = max(
                student_model[
                    "support_level"
                ],
                2
            )

            student_model[
                "next_action"
            ] = "DIAGNOSE"


        elif struggle == 2:

            student_model[
                "support_level"
            ] = 3

            student_model[
                "phase"
            ] = "GUIDED_PRACTICE"

            student_model[
                "next_action"
            ] = "CONTRAST"


        elif struggle == 3:

            student_model[
                "support_level"
            ] = 4

            student_model[
                "phase"
            ] = "GUIDED_PRACTICE"

            student_model[
                "next_action"
            ] = "HINT"


        elif struggle == 4:

            student_model[
                "support_level"
            ] = 5

            student_model[
                "phase"
            ] = "GUIDED_PRACTICE"

            student_model[
                "next_action"
            ] = "EXAMPLE"


        else:

            student_model[
                "support_level"
            ] = 6

            student_model[
                "phase"
            ] = "GUIDED_PRACTICE"

            student_model[
                "next_action"
            ] = "EXPLAIN"


        student_model[
            "decision_reason"
        ] = (
            "The answer is incorrect, but a stable "
            "misconception has not yet been established."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student's next response reveals "
            "their reasoning or improved understanding."
        )


    # ============================================================
    # MISCONCEPTION
    # ============================================================

    elif classification == "MISCONCEPTION":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "struggle_count"
        ] += 1

        student_model[
            "student_state"
        ] = "NEEDS_REVIEW"

        student_model[
            "phase"
        ] = "GUIDED_PRACTICE"


        student_model = (
            activate_misconception(
                student_model,
                reason
            )
        )


        if previous_action in [

            "PROBE",

            "NARROW",

            "DIAGNOSE"

        ]:

            student_model[
                "support_level"
            ] = max(
                student_model[
                    "support_level"
                ],
                3
            )

            student_model[
                "next_action"
            ] = "CONTRAST"


        elif previous_action == "CONTRAST":

            student_model[
                "support_level"
            ] = 4

            student_model[
                "next_action"
            ] = "HINT"


        elif previous_action == "HINT":

            student_model[
                "support_level"
            ] = 5

            student_model[
                "next_action"
            ] = "EXAMPLE"


        else:

            student_model[
                "support_level"
            ] = 6

            student_model[
                "next_action"
            ] = "EXPLAIN"


        student_model[
            "decision_reason"
        ] = (
            "The response demonstrates a specific "
            "incorrect mental model, so the support "
            "must directly target that model."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student revises the incorrect idea "
            "using scientifically correct reasoning."
        )


    # ============================================================
    # VOCABULARY ISSUE
    # ============================================================

    elif classification == "VOCABULARY_ISSUE":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "struggle_count"
        ] += 1

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "phase"
        ] = "GUIDED_PRACTICE"

        student_model[
            "support_level"
        ] = max(
            student_model[
                "support_level"
            ],
            5
        )

        student_model[
            "next_action"
        ] = "EXAMPLE"

        student_model[
            "decision_reason"
        ] = (
            "The main barrier is vocabulary rather "
            "than demonstrated conceptual failure."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student demonstrates understanding "
            "of the word and returns to the concept."
        )


    # ============================================================
    # PREREQUISITE GAP
    # ============================================================

    elif classification == "PREREQUISITE_GAP":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "struggle_count"
        ] += 1

        student_model[
            "student_state"
        ] = "NEEDS_REVIEW"

        student_model[
            "phase"
        ] = "GUIDED_PRACTICE"

        student_model[
            "support_level"
        ] = max(
            student_model[
                "support_level"
            ],
            5
        )

        student_model[
            "next_action"
        ] = "REMEDIATE_PREREQUISITE"

        student_model[
            "decision_reason"
        ] = (
            "The student is missing foundational "
            "knowledge required for the current objective."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student demonstrates the prerequisite "
            "idea before returning to the original objective."
        )


    # ============================================================
    # UNCERTAIN
    # ============================================================

    elif classification == "UNCERTAIN":

        student_model[
            "success_count"
        ] = 0

        student_model[
            "struggle_count"
        ] += 1

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "phase"
        ] = "DIAGNOSIS"


        struggle = student_model[
            "struggle_count"
        ]


        if struggle == 1:

            student_model[
                "support_level"
            ] = max(
                student_model[
                    "support_level"
                ],
                2
            )

            student_model[
                "next_action"
            ] = "NARROW"


        elif struggle == 2:

            student_model[
                "support_level"
            ] = max(
                student_model[
                    "support_level"
                ],
                3
            )

            student_model[
                "next_action"
            ] = "CONTRAST"


        elif struggle == 3:

            student_model[
                "support_level"
            ] = max(
                student_model[
                    "support_level"
                ],
                4
            )

            student_model[
                "next_action"
            ] = "HINT"


        elif struggle == 4:

            student_model[
                "support_level"
            ] = max(
                student_model[
                    "support_level"
                ],
                5
            )

            student_model[
                "next_action"
            ] = "EXAMPLE"


        else:

            student_model[
                "support_level"
            ] = 6

            student_model[
                "next_action"
            ] = "EXPLAIN"


        student_model[
            "decision_reason"
        ] = (
            "The student is uncertain, so make the "
            "thinking step smaller without assuming "
            "a misconception."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student makes a clearer prediction "
            "or identifies what is confusing."
        )


    # ============================================================
    # IRRELEVANT
    # ============================================================

    elif classification == "IRRELEVANT":

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "phase"
        ] = "DIAGNOSIS"

        student_model[
            "next_action"
        ] = "PROBE"

        student_model[
            "decision_reason"
        ] = (
            "The response does not provide evidence "
            "for the current objective."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student reconnects their response "
            "to the current concept."
        )


    # ============================================================
    # UNKNOWN
    # ============================================================

    else:

        student_model[
            "student_state"
        ] = "DEVELOPING"

        student_model[
            "phase"
        ] = "DIAGNOSIS"

        student_model[
            "support_level"
        ] = max(
            student_model[
                "support_level"
            ],
            2
        )

        student_model[
            "next_action"
        ] = "DIAGNOSE"

        student_model[
            "decision_reason"
        ] = (
            "The classifier result is unclear, "
            "so collect more evidence."
        )

        student_model[
            "expected_next_evidence"
        ] = (
            "The student provides enough information "
            "for a clearer diagnosis."
        )


    # ============================================================
    # DETERMINISTIC ADVANCEMENT GATE
    # ============================================================

    student_model[
        "advance_allowed"
    ] = can_advance(
        student_model
    )


    if (

        student_model[
            "next_action"
        ] == "ADVANCE"

        and not student_model[
            "advance_allowed"
        ]

    ):

        evidence = student_model[
            "evidence_matrix"
        ]


        if evidence[
            "explanation"
        ] != "PASSED":

            student_model[
                "phase"
            ] = "DIAGNOSIS"

            student_model[
                "next_action"
            ] = "PROBE"

            student_model[
                "expected_next_evidence"
            ] = (
                "The student explains the scientific "
                "reasoning correctly."
            )


        elif evidence[
            "independent_retest"
        ] != "PASSED":

            student_model[
                "phase"
            ] = "RETEST"

            student_model[
                "next_action"
            ] = "RETEST"

            student_model[
                "expected_next_evidence"
            ] = (
                "The student independently answers "
                "a fresh case without support."
            )


        elif evidence[
            "transfer"
        ] != "PASSED":

            student_model[
                "phase"
            ] = "TRANSFER"

            student_model[
                "next_action"
            ] = "TRANSFER"

            student_model[
                "expected_next_evidence"
            ] = (
                "The student independently solves "
                "a changed-context inference problem."
            )


        student_model[
            "decision_reason"
        ] = (
            "Advancement is blocked because required "
            "learning evidence is still missing."
        )


    return student_model


# ================================================================
# 12. DEBUG PANEL
# ================================================================

def print_debug(
    student_model,
    student_response
):

    print("\n" + "=" * 70)

    print("V3.1 TUTOR DEBUG")

    print("=" * 70)


    print(
        "CURRENT CONCEPT:",
        student_model["concept"]
    )

    print(
        "LEARNING OBJECTIVE:",
        learning_objective
    )

    print(
        "STUDENT RESPONSE:",
        student_response
    )

    print(
        "CLASSIFICATION:",
        student_model[
            "last_classification"
        ]
    )

    print(
        "STUDENT STATE:",
        student_model[
            "student_state"
        ]
    )

    print(
        "PHASE:",
        student_model[
            "phase"
        ]
    )

    print(
        "SUPPORT LEVEL:",
        student_model[
            "support_level"
        ]
    )

    print(
        "DIFFICULTY:",
        student_model[
            "difficulty"
        ]
    )

    print(
        "MISCONCEPTION:",
        student_model[
            "misconception"
        ]
    )

    print(
        "MISCONCEPTION STATUS:",
        student_model[
            "misconception_status"
        ]
    )

    print(
        "NEXT ACTION:",
        student_model[
            "next_action"
        ]
    )

    print(
        "DECISION REASON:",
        student_model[
            "decision_reason"
        ]
    )

    print(
        "EXPECTED NEXT EVIDENCE:",
        student_model[
            "expected_next_evidence"
        ]
    )

    print(
        "RETEST PASSED:",
        student_model[
            "retest_passed"
        ]
    )

    print(
        "TRANSFER PASSED:",
        student_model[
            "transfer_passed"
        ]
    )


    print("\nEVIDENCE MATRIX:")

    for evidence_type, status in (
        student_model[
            "evidence_matrix"
        ].items()
    ):

        print(
            f"  {evidence_type}: {status}"
        )


    print(
        "\nADVANCE ALLOWED:",
        student_model[
            "advance_allowed"
        ]
    )


    print("=" * 70)


# ================================================================
# 13. TUTOR RESPONSE GENERATOR
# ================================================================

def generate_tutor_response(
    student_model,
    classification_result,
    student_message,
    history_string
):

    prompt = f"""

You are the natural-language tutoring layer
of an adaptive Class 6 Science AI Tutor.

You are NOT a generic chatbot.

The Python Tutor Engine has already decided
the student's current learning state
and selected the next teaching action.

Your job is to express that decision naturally.


================================================
CURRENT LEARNING STATE
================================================

CONCEPT:

{student_model["concept"]}


LEARNING OBJECTIVE:

{learning_objective}


STUDENT STATE:

{student_model["student_state"]}


PHASE:

{student_model["phase"]}


SUPPORT LEVEL:

{student_model["support_level"]}


DIFFICULTY:

{student_model["difficulty"]}


MISCONCEPTION:

{student_model["misconception"]}


MISCONCEPTION STATUS:

{student_model["misconception_status"]}


EVIDENCE MATRIX:

{student_model["evidence_matrix"]}


SELECTED ACTION:

{student_model["next_action"]}


EXPECTED NEXT EVIDENCE:

{student_model["expected_next_evidence"]}


ADVANCE ALLOWED:

{student_model["advance_allowed"]}


CLASSIFIER RESULT:

{classification_result}


CURRENT STUDENT MESSAGE:

{student_message}


RECENT CONVERSATION:

{history_string}


================================================
CORE PURPOSE
================================================

Your purpose is NOT to do all the student's
thinking for them.

Help the student perform the next useful
piece of thinking.

Use the minimum useful support.

When the learner succeeds,
reduce support.

The scientific learning cycle may involve:

SEE
→
NOTICE
→
WONDER
→
ASK
→
PREDICT
→
REASON
→
CHECK
→
OBSERVE
→
COMPARE
→
EXPLAIN
→
CONNECT
→
QUESTION AGAIN


Science is not only about finding answers.

It is also about learning
how to ask better questions.


================================================
ONE MAIN QUESTION
================================================

Ask only ONE main question at a time.

Do not overload the learner
with several questions.

Usually use:

one or two useful thinking steps
→
clear explanation when necessary
→
one deeper related question.


================================================
FOLLOW THE ENGINE ACTION
================================================

Follow SELECTED ACTION.

Do not expose:

classification,

support level,

phase,

student state,

evidence matrix,

misconception status,

ADVANCE_ALLOWED,

or internal decision logic

to the student.


================================================
DIAGNOSE
================================================

If action is DIAGNOSE:

Find out WHY the student is struggling.

Distinguish between:

misconception,

vocabulary problem,

prerequisite gap,

uncertainty,

question misunderstanding,

or incomplete reasoning.

Ask ONE useful diagnostic question.


================================================
PROBE
================================================

If action is PROBE:

Ask for reasoning.

Examples:

"Why do you think that?"

"What rule helped you decide?"

"What made you choose that answer?"

Do not provide unnecessary explanation.


================================================
NARROW
================================================

If action is NARROW:

Make the thinking step smaller.

Ask specifically for the important
missing evidence.

Do not reveal the answer.


================================================
CONTRAST
================================================

If action is CONTRAST:

Give two meaningful alternatives
or models to compare.

The contrast MUST target
the specific misconception.

Do not use a generic contrast
when a specific incorrect mental model
has already been identified.


Example:

If misconception:

"North and south are positive
and negative electrical charges."

Do NOT merely ask:

"What happens with N-N and N-S?"


Instead contrast:

A. North and south are positive
   and negative electrical charges.

B. North and south are magnetic poles,
   not positive and negative
   electrical charges.


Ask the student to compare
or choose and explain.

Do not reveal the answer immediately
unless the support level requires it.


================================================
HINT
================================================

If action is HINT:

Give ONE small clue.

Do not reveal the complete answer.


================================================
EXAMPLE
================================================

If action is EXAMPLE:

Give ONE simple relevant example.

If it is an analogy,
explicitly make clear that it is an analogy
when necessary.

Do not create a false scientific model.


================================================
EXPLAIN
================================================

If action is EXPLAIN:

Give a concise scientifically correct explanation.

Use:

simple idea
→
correct scientific term
→
short example

Then prepare the learner
for an independent retest.


================================================
REMEDIATE_PREREQUISITE
================================================

If action is REMEDIATE_PREREQUISITE:

Teach ONLY the minimum foundation
needed for the current objective.

Use:

FOUNDATION
→
SIMPLE MEANING
→
ONE EXAMPLE
→
CHECK UNDERSTANDING
→
RETURN TO ORIGINAL OBJECTIVE


For magnetic poles,
the minimum foundation may include:

1. A magnet can exert magnetic force.

2. A magnet has two poles:
   north and south.

3. North and south are magnetic poles.

4. They are NOT positive and negative
   electrical charges.


Do not demand an advanced prediction
before checking that the prerequisite
is understood.


================================================
RETEST
================================================

If action is RETEST:

Remove support.

No hints.

No choices.

Do not include the answer.

Ask a fresh case independently.


Example:

"Now try this without a hint:
What happens when two south poles
are brought close together?
Explain why."


================================================
TRANSFER
================================================

If action is TRANSFER:

Test the SAME concept
in a changed situation.

Transfer must require inference,
not simple repetition.


For the Magnetic Poles benchmark,
prefer this type of problem:

"Magnet A and Magnet B are facing each other.

They push away from each other.

Magnet A's facing pole is North.

What must Magnet B's facing pole be?

Explain how you know."


Expected scientific reasoning:

Magnet B must also be North
because like poles repel.


Do NOT reveal that expected answer
inside the question.


Do NOT count simply repeating:

"Like poles repel and unlike poles attract"

as sufficient transfer.


Do NOT suddenly change the concept to:

paperclips,

magnetic materials,

electromagnets,

compasses,

or magnetic fields.


Transfer means:

SAME CONCEPT
+
NEW SITUATION

not:

NEW CONCEPT.


================================================
ADVANCE
================================================

If action is ADVANCE:

Advance ONLY when:

ADVANCE ALLOWED = True.


If ADVANCE ALLOWED = False:

DO NOT advance.


Never override the Python progression gate
because the learner sounds confident.


A single correct answer
is not mastery.

A correct answer after a hint
is not independent mastery.

Transfer must actually be demonstrated.


================================================
MISCONCEPTION LIFECYCLE
================================================

Possible statuses:

ACTIVE

REDUCED

RESOLVED


ACTIVE:

Evidence currently supports
the incorrect mental model.


REDUCED:

The learner improved after targeted support,
but sufficient independent evidence
does not yet exist.


RESOLVED:

Independent retest
AND changed-context transfer
support the corrected model.


Never imply that a misconception is resolved
because the student:

said "okay",

copied the tutor,

or answered one guided question correctly.


================================================
SCIENTIFIC ACCURACY
================================================

Simplify LANGUAGE.

Do NOT distort SCIENCE.


Do NOT say:

"Magnets want to come together."


Prefer:

"Unlike magnetic poles attract."


Do NOT say:

"All metals are magnetic."


Prefer:

"Only some materials,
such as iron,
are strongly attracted
to ordinary magnets."


Do NOT say:

"Field lines physically push magnets."


If field lines are discussed:

"Magnetic field lines are a representation
used to show the direction and pattern
of a magnetic field."


North and south magnetic poles
are NOT positive and negative
electrical charges.


If using a + / - analogy,
explicitly state that it is only an analogy.


================================================
PREDICTION VS OBSERVATION
================================================

Prediction:

What the learner thinks will happen.


Observation:

What the learner actually reports seeing.


Evidence:

Information supporting
or challenging an idea.


Explanation:

Scientific reasoning
used to understand the result.


Never pretend an experiment happened
if the student did not report doing it.


================================================
SAFE SCIENCE
================================================

Never instruct a Class 6 learner to:

touch sockets,

open electrical appliances,

use mains electricity,

look directly at the Sun,

mix unknown chemicals,

taste unknown substances,

handle flames independently,

touch dangerously hot objects,

or climb unsafe places.


Prefer low-risk classroom
or household observations.


================================================
QUESTION QUALITY
================================================

Gradually help the learner move from:

WHAT?

to

HOW?

to

WHY?

to

WHAT IF?

to

HOW COULD WE TEST IT?

to

WHAT EVIDENCE DO WE HAVE?


Do not force this sequence mechanically.


================================================
FEEDBACK
================================================

Use specific feedback.

Good:

"Correct — you used the rule
that like poles repel."


Avoid generic overpraise such as:

"Amazing!"

"Perfect!"

"Brilliant!"


If the answer is incomplete,
do not say it is fully correct.


================================================
STYLE
================================================

Use natural Class 6 language.

Do not sound like a research paper.

Do not artificially speak like a baby.

Normally respond in about 30–100 words.

Use one main question.

Keep the lesson anchored
to the current learning objective.


================================================
FINAL RULE
================================================

The goal is not merely:

QUESTION
→
ANSWER

The goal is:

STUDENT RESPONSE
→
EVIDENCE
→
DECISION
→
MINIMUM SUPPORT
→
INDEPENDENT RETEST
→
TRANSFER
→
ADVANCEMENT ONLY WHEN JUSTIFIED.

"""

    response = client.chat.completions.create(

        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],

        model=MODEL_NAME
    )

    return response.choices[0].message.content


# ================================================================
# 14. V3.1 FIXED VALIDATION BENCHMARK
# ================================================================

benchmark_cases = [

    {
        "id": "T01",

        "response":
            "North and north repel.",

        "expected_classification":
            ["CORRECT"],

        "expected_state":
            ["DEVELOPING"],

        "expected_misconception_status":
            [None],

        "expected_action":
            ["DIAGNOSE", "PROBE"],

        "expected_advance":
            False,

        "validation_note":
            "Correct does not equal mastery."
    },


    {
        "id": "T02",

        "response":
            "North and north attract because both are magnets.",

        "expected_classification":
            ["MISCONCEPTION"],

        "expected_state":
            ["NEEDS_REVIEW"],

        "expected_misconception_status":
            ["ACTIVE"],

        "expected_action":
            ["CONTRAST"],

        "expected_advance":
            False,

        "validation_note":
            (
                "The incorrect general model must be "
                "targeted rather than merely marking "
                "the final answer wrong."
            )
    },


    {
        "id": "T03",

        "response":
            "I don't know.",

        "expected_classification":
            ["UNCERTAIN"],

        "expected_state":
            ["DEVELOPING"],

        "expected_misconception_status":
            [None],

        "expected_action":
            [
                "PROBE",
                "NARROW",
                "HINT"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "Uncertainty alone must not create "
                "a misconception."
            )
    },


    {
        "id": "T04",

        "response":
            "Like poles repel because they are different.",

        "expected_classification":
            [
                "MISCONCEPTION",
                "PARTIALLY_CORRECT"
            ],

        "expected_state":
            [
                "NEEDS_REVIEW",
                "DEVELOPING"
            ],

        "expected_misconception_status":
            [
                "ACTIVE",
                None
            ],

        "expected_action":
            [
                "CONTRAST",
                "NARROW"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "The final conclusion is correct, "
                "but the reasoning is conceptually wrong."
            )
    },


    {
        "id": "T05",

        "response":
            "They attract.",

        "expected_classification":
            [
                "INSUFFICIENT_EVIDENCE",
                "PARTIALLY_CORRECT"
            ],

        "expected_state":
            ["DEVELOPING"],

        "expected_misconception_status":
            [None],

        "expected_action":
            [
                "PROBE",
                "NARROW"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "If pole identity and reasoning were "
                "requested, 'They attract' is insufficient."
            )
    },


    {
        "id": "T06",

        "response":
            (
                "North and south attract because "
                "unlike poles attract."
            ),

        "expected_classification":
            ["CORRECT"],

        "expected_state":
            ["DEVELOPING"],

        "expected_misconception_status":
            [None],

        "expected_action":
            [
                "RETEST",
                "PROBE"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "One correct response must not "
                "immediately cause advancement."
            )
    },


    {
        "id": "T07",

        "response":
            "Correct response after tutor support.",

        "expected_classification":
            ["CORRECT"],

        "expected_state":
            ["DEVELOPING"],

        "expected_misconception_status":
            [
                "ACTIVE",
                "REDUCED",
                None
            ],

        "expected_action":
            ["RETEST"],

        "expected_advance":
            False,

        "validation_note":
            (
                "Correctness after support requires "
                "an independent retest."
            )
    },


    {
        "id": "T08",

        "response":
            "Student fails the independent retest.",

        "expected_classification":
            [
                "INCORRECT",
                "MISCONCEPTION"
            ],

        "expected_state":
            ["NEEDS_REVIEW"],

        "expected_misconception_status":
            [
                "ACTIVE",
                "REDUCED",
                None
            ],

        "expected_action":
            [
                "DIAGNOSE",
                "CONTRAST",
                "HINT",
                "EXAMPLE",
                "EXPLAIN"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "A failed independent retest "
                "must block advancement."
            )
    },


    {
        "id": "T09",

        "response":
            (
                "North + North = repel; "
                "North + South = attract, "
                "because like poles repel "
                "and unlike poles attract."
            ),

        "expected_classification":
            ["CORRECT"],

        "expected_state":
            ["DEVELOPING"],

        "expected_misconception_status":
            [
                None,
                "REDUCED"
            ],

        "expected_action":
            ["TRANSFER"],

        "expected_advance":
            False,

        "validation_note":
            (
                "Correct rule use should lead "
                "to changed-context transfer."
            )
    },


    {
        "id": "T10",

        "response":
            (
                "Magnet A is North. "
                "B pushes away, so B is South."
            ),

        "expected_classification":
            ["MISCONCEPTION"],

        "expected_state":
            ["NEEDS_REVIEW"],

        "expected_misconception_status":
            ["ACTIVE"],

        "expected_action":
            [
                "CONTRAST",
                "PROBE"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "The student failed the "
                "changed-context inference."
            )
    },


    {
        "id": "T11",

        "response":
            (
                "B must be North because "
                "like poles repel."
            ),

        "expected_classification":
            ["CORRECT"],

        "expected_state":
            [
                "DEVELOPING",
                "SECURE"
            ],

        "expected_misconception_status":
            [
                None,
                "RESOLVED"
            ],

        "expected_action":
            [
                "ADVANCE",
                "PROBE"
            ],

        "expected_advance":
            True,

        "validation_note":
            (
                "Advance only when all other "
                "required evidence is already satisfied."
            )
    },


    {
        "id": "T12",

        "response":
            (
                "North is positive and "
                "south is negative."
            ),

        "expected_classification":
            ["MISCONCEPTION"],

        "expected_state":
            ["NEEDS_REVIEW"],

        "expected_misconception_status":
            ["ACTIVE"],

        "expected_action":
            ["CONTRAST"],

        "expected_advance":
            False,

        "validation_note":
            (
                "Magnetic poles must be separated "
                "from positive and negative "
                "electrical charges."
            )
    },


    {
        "id": "T13",

        "response":
            (
                "Field lines physically push "
                "the magnets apart."
            ),

        "expected_classification":
            ["MISCONCEPTION"],

        "expected_state":
            ["NEEDS_REVIEW"],

        "expected_misconception_status":
            ["ACTIVE"],

        "expected_action":
            [
                "CONTRAST",
                "EXPLAIN"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "Field lines are representations, "
                "not physical objects that push magnets."
            )
    },


    {
        "id": "T14",

        "response":
            (
                "Student succeeds after a hint "
                "but fails an unseen transfer case."
            ),

        "expected_classification":
            [
                "INCORRECT",
                "MISCONCEPTION"
            ],

        "expected_state":
            ["NEEDS_REVIEW"],

        "expected_misconception_status":
            [
                "ACTIVE",
                "REDUCED",
                None
            ],

        "expected_action":
            [
                "DIAGNOSE",
                "CONTRAST",
                "HINT",
                "EXAMPLE",
                "EXPLAIN"
            ],

        "expected_advance":
            False,

        "validation_note":
            (
                "Guided success does not compensate "
                "for failed transfer."
            )
    },


    {
        "id": "T15",

        "response":
            (
                "Student independently solves "
                "an unseen magnetic-pole inference "
                "and correctly explains why."
            ),

        "expected_classification":
            ["CORRECT"],

        "expected_state":
            [
                "SECURE",
                "DEVELOPING"
            ],

        "expected_misconception_status":
            [
                None,
                "RESOLVED"
            ],

        "expected_action":
            [
                "ADVANCE"
            ],

        "expected_advance":
            True,

        "validation_note":
            (
                "Advance only when the full required "
                "evidence matrix is satisfied."
            )
    }
]


# ================================================================
# 15. BENCHMARK RESULT CHECKER
# ================================================================

def validate_benchmark_result(
    case,
    actual_result
):

    checks = {}


    checks["classification"] = (

        actual_result[
            "classification"
        ]

        in case[
            "expected_classification"
        ]
    )


    checks["state"] = (

        actual_result[
            "state"
        ]

        in case[
            "expected_state"
        ]
    )


    checks[
        "misconception_status"
    ] = (

        actual_result[
            "misconception_status"
        ]

        in case[
            "expected_misconception_status"
        ]
    )


    checks["action"] = (

        actual_result[
            "action"
        ]

        in case[
            "expected_action"
        ]
    )


    checks[
        "advance_allowed"
    ] = (

        actual_result[
            "advance_allowed"
        ]

        == case[
            "expected_advance"
        ]
    )


    passed = all(
        checks.values()
    )


    print("\n" + "=" * 70)

    print(
        case["id"],
        "VALIDATION"
    )

    print("=" * 70)


    print(
        "STUDENT RESPONSE:"
    )

    print(
        case["response"]
    )


    print("\nRESULTS:")


    for name, result in checks.items():

        print(
            f"{name.upper():25}",
            "PASS"
            if result
            else "FAIL"
        )


    print(
        "\nVALIDATION NOTE:"
    )

    print(
        case[
            "validation_note"
        ]
    )


    print(
        "\nFINAL RESULT:",
        "PASS"
        if passed
        else "FAIL"
    )


    print("=" * 70)


    return passed


# ================================================================
# 16. CREATE DEBUG RESULT
# ================================================================

def get_debug_result(
    student_model
):

    return {

        "classification":
            student_model[
                "last_classification"
            ],

        "state":
            student_model[
                "student_state"
            ],

        "misconception_status":
            student_model[
                "misconception_status"
            ],

        "action":
            student_model[
                "next_action"
            ],

        "advance_allowed":
            student_model[
                "advance_allowed"
            ]
    }


# ================================================================
# 17. SHOW BENCHMARK SPECIFICATION
# ================================================================

def show_benchmark_cases():

    print("\n" + "=" * 70)

    print(
        "V3.1 VALIDATION BENCHMARK"
    )

    print("=" * 70)


    for case in benchmark_cases:

        print(
            "\n",
            case["id"],
            sep=""
        )

        print(
            "Response:",
            case["response"]
        )

        print(
            "Expected Classification:",
            case[
                "expected_classification"
            ]
        )

        print(
            "Expected State:",
            case[
                "expected_state"
            ]
        )

        print(
            "Expected Misconception Status:",
            case[
                "expected_misconception_status"
            ]
        )

        print(
            "Expected Action:",
            case[
                "expected_action"
            ]
        )

        print(
            "Expected Advance:",
            case[
                "expected_advance"
            ]
        )

        print(
            "Validation Note:",
            case[
                "validation_note"
            ]
        )

        print("-" * 70)


# ================================================================
# 18. GATE REPORT
# ================================================================

def print_gate_report(
    gate_results
):

    print("\n" + "=" * 70)

    print(
        "V3.1 VALIDATION GATE REPORT"
    )

    print("=" * 70)


    gates = [

        "G1 Classification",

        "G2 Misconception Lifecycle",

        "G3 Evidence Accumulation",

        "G4 Expected Evidence Gating",

        "G5 Support vs Phase",

        "G6 Independent Retest",

        "G7 Transfer",

        "G8 Advancement Control",

        "G9 Adversarial Reliability",

        "G10 Scientific Accuracy"
    ]


    for gate in gates:

        result = gate_results.get(
            gate,
            "NOT_TESTED"
        )

        print(
            f"{gate}: {result}"
        )


    print("=" * 70)


# ================================================================
# 19. CONVERSATION MEMORY
# ================================================================

history_turns = []

MEMORY_LIMIT = 20

last_tutor_response = ""


# ================================================================
# 20. MAIN TUTOR LOOP
# ================================================================

while True:

    question = input(
        "\nStudent: "
    )


    if (
        question
        .strip()
        .lower()
        == "exit"
    ):

        print("\nTutor:")

        print(
            "Goodbye! Keep observing, "
            "wondering, and asking "
            "better questions."
        )

        break


    if not question.strip():

        continue


    history_string = "\n\n".join(
        history_turns
    )


    # ------------------------------------------------------------
    # CLASSIFICATION
    # ------------------------------------------------------------

    if last_tutor_response:

        classification_result = (
            classify_student_response(

                current_concept=
                    current_concept,

                learning_objective=
                    learning_objective,

                tutor_question=
                    last_tutor_response,

                student_response=
                    question,

                history_string=
                    history_string,

                expected_next_evidence=
                    student_model[
                        "expected_next_evidence"
                    ]
            )
        )

    else:

        classification_result = (

            "CLASSIFICATION: INITIAL_QUERY\n"

            "REASON: "
            "This is the first student message."
        )


    # ------------------------------------------------------------
    # UPDATE STUDENT MODEL
    # ------------------------------------------------------------

    student_model = update_tutor_state(

        classification_result,

        student_model
    )


    # ------------------------------------------------------------
    # DEBUG PANEL
    # ------------------------------------------------------------

    print_debug(

        student_model,

        question
    )


    # ------------------------------------------------------------
    # GENERATE TUTOR RESPONSE
    # ------------------------------------------------------------

    tutor_response = (
        generate_tutor_response(

            student_model,

            classification_result,

            question,

            history_string
        )
    )


    print("\nTutor:")

    print(
        tutor_response
    )


    # ------------------------------------------------------------
    # UPDATE CONVERSATION MEMORY
    # ------------------------------------------------------------

    history_turns.append(

        "Student:\n"
        + question
        + "\n\nTutor:\n"
        + tutor_response
    )


    if len(
        history_turns
    ) > MEMORY_LIMIT:

        history_turns.pop(0)


    last_tutor_response = (
        tutor_response
    )
```

```python
import os
from getpass import getpass
from groq import Groq

MODEL_NAME = "openai/gpt-oss-120b"


# Change this to:"TUTOR"for normal tutoring.
# Change to:"VALIDATION", to run the full V3.1 Validation Gate.

MODE = "TUTOR"
REPEATABILITY_RUNS = 3


os.environ["GROQ_API_KEY"] = getpass(
    "Enter your Groq API key: "
)


client = Groq(
    api_key=os.environ["GROQ_API_KEY"]
)


current_concept = "Magnetic poles"


learning_objective = (
    "Understand that like poles repel "
    "and unlike poles attract."
)

SUPPORTED_ACTIONS = [

    "NARROW",

    "CONTRAST",

    "HINT",

    "EXAMPLE",

    "EXPLAIN",

    "CLARIFY",

    "REMEDIATE_PREREQUISITE"
]


KNOWN_MISCONCEPTION_IDS = [

    "LIKE_POLES_ATTRACT",

    "MAGNETS_ALWAYS_ATTRACT",

    "LIKE_MEANS_DIFFERENT",

    "MAGNETIC_POLES_ARE_ELECTRIC_CHARGES",

    "FIELD_LINES_ARE_PHYSICAL_OBJECTS",

    "TRANSFER_RULE_REVERSAL",

    "OTHER"
]


def create_student_model():

    return {

        "concept":
            current_concept,

        "student_state":
            "NOT_STARTED",

        "support_level":
            1,

        "phase":
            "DIAGNOSIS",

        "misconception_id":
            None,

        "misconception":
            None,

        "misconception_status":
            None,

        "evidence":
            [],

        "evidence_matrix": {

            "prior_knowledge":
                "PENDING",

            "prediction":
                "PENDING",

            "explanation":
                "PENDING",

            "guided_application":
                "PENDING",

            "independent_retest":
                "PENDING",

            "transfer":
                "PENDING"
        },

        "difficulty":
            "Medium",

        "next_action":
            "PROBE",

        "last_classification":
            None,

        "last_semantic_event":
            None,

        "success_count":
            0,

        "struggle_count":
            0,

        "retest_passed":
            False,

        "transfer_passed":
            False,

        "advance_allowed":
            False,

        "decision_reason":
            "",

        "expected_evidence_types": [

            "PRIOR_KNOWLEDGE",

            "PREDICTION"
        ],

        "expected_next_evidence":
            (
                "Reveal prior knowledge "
                "or make an initial prediction."
            )
    }


student_model = create_student_model()

def classify_student_response(

    current_concept,

    learning_objective,

    tutor_question,

    student_response,

    history_string,

    expected_evidence_types,

    expected_next_evidence
):


    classifier_prompt = f"""

You are the Student Response Classifier
inside a Class 6 Science AI Tutor.


Your ONLY job is to determine
what the student's response actually proves.


Do NOT tutor.

Do NOT praise.

Do NOT give hints.

Do NOT explain the lesson.

Do NOT decide mastery by yourself.

Do NOT invent evidence.


CURRENT CONCEPT:

{current_concept}


LEARNING OBJECTIVE:

{learning_objective}


TUTOR'S PREVIOUS QUESTION:

{tutor_question}


EXPECTED EVIDENCE TYPES:

{expected_evidence_types}


EXPECTED EVIDENCE DESCRIPTION:

{expected_next_evidence}


STUDENT RESPONSE:

{student_response}


RECENT CONVERSATION:

{history_string}


================================================
BEFORE CLASSIFYING
================================================

Silently check these questions:


1. What exactly did the Tutor ask?


2. What evidence was requested?


3. What exactly did the student provide?


4. Did the student answer
   the actual question?


5. Is the conclusion
   scientifically correct?


6. Is the reasoning
   scientifically correct?


7. Is important evidence missing?


8. Does the response reveal
   a specific false mental model?


9. Is the main difficulty vocabulary?


10. Is earlier prerequisite knowledge missing?


11. Is the student simply uncertain?


12. Does the current response conflict
    with something the student said earlier?


================================================
ALLOWED CLASSIFICATIONS
================================================

Choose ONLY ONE:


CORRECT

PARTIALLY_CORRECT

INSUFFICIENT_EVIDENCE

INCORRECT

MISCONCEPTION

VOCABULARY_ISSUE

PREREQUISITE_GAP

UNCERTAIN

IRRELEVANT


================================================
CORRECT
================================================

Use CORRECT only when
the student correctly answers
the actual question

AND

gives the important evidence requested.


If reasoning was requested,
the reasoning must also be correct.


A scientifically true sentence
is NOT automatically
a correct answer
to the Tutor's question.


A correct answer
is NOT automatically mastery.


================================================
PARTIALLY_CORRECT
================================================

Use when part of the answer
or reasoning is useful,

but something important
is missing
or slightly incorrect.


Do NOT use PARTIALLY_CORRECT
when the response clearly demonstrates
a stable false mental model.


In that case:

MISCONCEPTION


================================================
INSUFFICIENT_EVIDENCE
================================================

Use when the response
may contain something true,

but does not provide
the evidence requested.


Example:


Tutor:

"An unknown pole attracts
a north pole.

What is the unknown pole
and why?"


Student:

"They attract."


The student did not identify
the unknown pole

and

did not explain why.


Classification:

INSUFFICIENT_EVIDENCE


================================================
INCORRECT
================================================

Use when the answer
is scientifically wrong

but there is not yet enough evidence
to identify a specific
stable misconception.


================================================
MISCONCEPTION
================================================

Use when the response demonstrates
a specific false scientific model,
rule,
or interpretation.


IMPORTANT PRIORITY RULE:


If the student gives
a correct final conclusion

but also demonstrates
a specific false mental model,


MISCONCEPTION


takes priority over


PARTIALLY_CORRECT.


Example:


"Like poles repel
because they are different."


The conclusion:

"Like poles repel"


is correct.


But the student believes:

"like means different."


That is a specific
incorrect conceptual model.


Therefore classify:

MISCONCEPTION


================================================
MISCONCEPTION EXAMPLES
================================================


Example 1:


"North and north attract
because both are magnets."


Classification:

MISCONCEPTION


------------------------------------------------


Example 2:


"Like poles repel
because they are different."


Classification:

MISCONCEPTION


------------------------------------------------


Example 3:


"North is positive
and south is negative."


Classification:

MISCONCEPTION


------------------------------------------------


Example 4:


"Field lines physically push
magnets apart."


Classification:

MISCONCEPTION


------------------------------------------------


Example 5:


Magnet A is North.

A and B repel.

Student says:

"B must be South."


Classification:

MISCONCEPTION


================================================
VOCABULARY ISSUE
================================================

Use when the main problem
is understanding a word.


Example:


"What does repel mean?"


================================================
PREREQUISITE GAP
================================================

Use PREREQUISITE_GAP only when
the student's response gives
clear evidence that an earlier
required idea is missing.


Examples:


"I don't know what a magnet is."


"I don't understand
what a magnetic pole means."


"I don't know anything about this.
Teach me from the basics."


"Can you first explain the basics?
I don't understand magnets yet."


IMPORTANT:


A student merely ASKING:


"What is a magnet?"


"What is a pole?"


"What does repel mean?"


does NOT automatically prove
that the learner has no prior knowledge.


At the beginning of a conversation,
first discover what the learner
already thinks.


Do NOT infer lack of knowledge
without evidence.


================================================
UNCERTAIN
================================================

Use when the student
does not know
or is unsure

but does NOT identify
a specific missing prerequisite.


Examples:


"I don't know."


"I'm not sure."


"Maybe."


================================================
IRRELEVANT
================================================

Use when the response
does not meaningfully address
the current science interaction.


================================================
LUCKY GUESS RULE
================================================

Suppose the Tutor asks:


"Magnet A is North.

A and B repel.

What must B be,

and why?"


Student says:


"North."


The final pole may be correct,

but the requested reasoning
is missing.


Do NOT count this
as complete transfer evidence.


Use:


INSUFFICIENT_EVIDENCE


or


PARTIALLY_CORRECT


================================================
MISCONCEPTION IDS
================================================

If classification is MISCONCEPTION,
choose ONE:


LIKE_POLES_ATTRACT


MAGNETS_ALWAYS_ATTRACT


LIKE_MEANS_DIFFERENT


MAGNETIC_POLES_ARE_ELECTRIC_CHARGES


FIELD_LINES_ARE_PHYSICAL_OBJECTS


TRANSFER_RULE_REVERSAL


OTHER


If classification
is NOT MISCONCEPTION:


MISCONCEPTION_ID must be:


NONE


================================================
SCIENCE
================================================

North and south magnetic poles
are NOT positive and negative
electrical charges.


Magnetic field lines
are representations
used to show a magnetic field.


They are NOT physical strings
or objects
that push magnets.


================================================
OUTPUT
================================================

Return EXACTLY:


CLASSIFICATION: <one allowed classification>

REASON: <one concise sentence>

MISCONCEPTION_ID: <one allowed ID or NONE>

MISCONCEPTION: <NONE or one short description>

"""


    response = client.chat.completions.create(

        messages=[
            {
                "role": "user",
                "content": classifier_prompt
            }
        ],

        model=MODEL_NAME,

        temperature=0
    )


    return response.choices[0].message.content


def parse_classification(
    classification_result
):


    classification = "UNKNOWN"

    reason = ""

    misconception_id = None

    misconception = None


    for raw_line in (
        classification_result
        .splitlines()
    ):


        line = raw_line.strip()


        if line.startswith(
            "CLASSIFICATION:"
        ):

            classification = (

                line.replace(
                    "CLASSIFICATION:",
                    ""
                )

                .strip()

                .upper()
            )


        elif line.startswith(
            "REASON:"
        ):

            reason = (

                line.replace(
                    "REASON:",
                    ""
                )

                .strip()
            )


        elif line.startswith(
            "MISCONCEPTION_ID:"
        ):

            value = (

                line.replace(
                    "MISCONCEPTION_ID:",
                    ""
                )

                .strip()

                .upper()
            )


            if value != "NONE":

                if value in (
                    KNOWN_MISCONCEPTION_IDS
                ):

                    misconception_id = value

                else:

                    misconception_id = "OTHER"


        elif line.startswith(
            "MISCONCEPTION:"
        ):

            value = (

                line.replace(
                    "MISCONCEPTION:",
                    ""
                )

                .strip()
            )


            if value.upper() != "NONE":

                misconception = value


    if classification != "MISCONCEPTION":

        misconception_id = None

        misconception = None


    return (

        classification,

        reason,

        misconception_id,

        misconception
    )


# ================================================================
# SUPPORT CONTROL
# ================================================================

def increase_support(
    model
):


    if model[
        "support_level"
    ] < 6:

        model[
            "support_level"
        ] += 1


    return model


# ================================================================
# EXPECTED EVIDENCE
# ================================================================

def set_expected(

    model,

    evidence_types,

    description
):


    model[
        "expected_evidence_types"
    ] = evidence_types


    model[
        "expected_next_evidence"
    ] = description


    return model

# ADVANCEMENT GATE
# ================================================================

def can_advance(
    model
):


    evidence = model[
        "evidence_matrix"
    ]


    return (

        evidence[
            "explanation"
        ] == "PASSED"

        and

        evidence[
            "independent_retest"
        ] == "PASSED"

        and

        evidence[
            "transfer"
        ] == "PASSED"
    )


# ================================================================
# EVIDENCE UPDATE
# ================================================================

def update_evidence_matrix(

    model,

    classification,

    previous_action,

    previous_phase,

    previous_expected_types
):


    evidence = model[
        "evidence_matrix"
    ]


    # ------------------------------------------------------------
    # PRIOR KNOWLEDGE
    # ------------------------------------------------------------

    if classification in [

        "CORRECT",

        "PARTIALLY_CORRECT",

        "INSUFFICIENT_EVIDENCE",

        "INCORRECT",

        "MISCONCEPTION"

    ]:


        evidence[
            "prior_knowledge"
        ] = "COLLECTED"


    # ------------------------------------------------------------
    # PREDICTION
    # ------------------------------------------------------------

    if (
        "PREDICTION"
        in previous_expected_types
    ):


        if classification == "CORRECT":


            evidence[
                "prediction"
            ] = "PASSED"


        elif classification in [

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "INCORRECT",

            "MISCONCEPTION"

        ]:


            evidence[
                "prediction"
            ] = "PARTIAL"


    # ------------------------------------------------------------
    # EXPLANATION
    # ------------------------------------------------------------

    if (
        "EXPLANATION"
        in previous_expected_types
    ):


        if classification == "CORRECT":


            evidence[
                "explanation"
            ] = "PASSED"


        elif classification in [

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "INCORRECT",

            "MISCONCEPTION"

        ]:


            evidence[
                "explanation"
            ] = "PARTIAL"


    # ------------------------------------------------------------
    # GUIDED APPLICATION
    # ------------------------------------------------------------

    if (

        previous_action
        in SUPPORTED_ACTIONS

        and

        classification
        == "CORRECT"

    ):


        evidence[
            "guided_application"
        ] = "PASSED"


    # ------------------------------------------------------------
    # INDEPENDENT RETEST
    # ------------------------------------------------------------

    if previous_phase == "RETEST":


        if classification == "CORRECT":


            evidence[
                "independent_retest"
            ] = "PASSED"


        elif classification in [

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "INCORRECT",

            "MISCONCEPTION",

            "UNCERTAIN"

        ]:


            evidence[
                "independent_retest"
            ] = "FAILED"


    # ------------------------------------------------------------
    # TRANSFER
    # ------------------------------------------------------------

    if previous_phase == "TRANSFER":


        if classification == "CORRECT":


            evidence[
                "transfer"
            ] = "PASSED"


        elif classification in [

            "PARTIALLY_CORRECT",

            "INSUFFICIENT_EVIDENCE",

            "INCORRECT",

            "MISCONCEPTION",

            "UNCERTAIN"

        ]:


            evidence[
                "transfer"
            ] = "FAILED"


    return model


# ================================================================
# MISCONCEPTION LIFECYCLE
# ================================================================

def activate_misconception(

    model,

    misconception_id,

    misconception,

    reason
):


    model[
        "misconception_id"
    ] = (

        misconception_id
        or
        "OTHER"
    )


    model[
        "misconception"
    ] = (

        misconception
        or
        reason
    )


    model[
        "misconception_status"
    ] = "ACTIVE"


    return model


def reduce_misconception(
    model
):


    if (

        model[
            "misconception_status"
        ]
        == "ACTIVE"

    ):


        model[
            "misconception_status"
        ] = "REDUCED"


    return model


def resolve_misconception_if_ready(
    model
):


    evidence = model[
        "evidence_matrix"
    ]


    if (

        model[
            "misconception_status"
        ]
        == "REDUCED"

        and

        evidence[
            "independent_retest"
        ]
        == "PASSED"

        and

        evidence[
            "transfer"
        ]
        == "PASSED"

    ):


        model[
            "misconception_status"
        ] = "RESOLVED"


    return model


# ================================================================
# SEMANTIC EVENT
# ================================================================

def derive_semantic_event(

    classification,

    previous_action,

    previous_phase,

    model
):


    evidence = model[
        "evidence_matrix"
    ]


    if previous_phase == "TRANSFER":


        if evidence[
            "transfer"
        ] == "PASSED":


            return "TRANSFER_PASSED"


        if evidence[
            "transfer"
        ] == "FAILED":


            return "FAILED_TRANSFER"


    if previous_phase == "RETEST":


        if evidence[
            "independent_retest"
        ] == "PASSED":


            return "RETEST_PASSED"


        if evidence[
            "independent_retest"
        ] == "FAILED":


            return "FAILED_RETEST"


    if (

        classification
        == "CORRECT"

        and

        previous_action
        in SUPPORTED_ACTIONS

    ):


        return "CORRECT_AFTER_SUPPORT"


    return classification


# ================================================================
# DECISION ENGINE
# ================================================================

def update_tutor_state(

    classification_result,

    model
):


    (

        classification,

        reason,

        misconception_id,

        misconception

    ) = parse_classification(
        classification_result
    )


    previous_action = model[
        "next_action"
    ]


    previous_phase = model[
        "phase"
    ]


    previous_expected_types = list(

        model[
            "expected_evidence_types"
        ]
    )


    model[
        "last_classification"
    ] = classification


    # ============================================================
    # INITIAL QUERY
    # ============================================================

    if classification == "INITIAL_QUERY":


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "support_level"
        ] = 1


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        model[
            "decision_reason"
        ] = (
            "This is the beginning of the interaction, "
            "so first reveal what the learner already thinks."
        )


        set_expected(

            model,

            [
                "PRIOR_KNOWLEDGE",

                "PREDICTION"
            ],

            (
                "The learner reveals prior knowledge, "
                "a guess, or an initial mental model."
            )
        )


        model[
            "advance_allowed"
        ] = False


        model[
            "last_semantic_event"
        ] = "INITIAL_QUERY"


        return model


    # ============================================================
    # UPDATE EVIDENCE
    # ============================================================

    model = update_evidence_matrix(

        model,

        classification,

        previous_action,

        previous_phase,

        previous_expected_types
    )


    model[
        "evidence"
    ].append(

        f"{classification}: {reason}"
    )


    if len(
        model[
            "evidence"
        ]
    ) > 20:


        model[
            "evidence"
        ].pop(0)


    # ============================================================
    # CORRECT
    # ============================================================

    if classification == "CORRECT":


        model[
            "success_count"
        ] += 1


        model[
            "struggle_count"
        ] = 0


        # --------------------------------------------------------
        # CORRECT DURING RETEST
        # --------------------------------------------------------

        if previous_phase == "RETEST":


            model[
                "student_state"
            ] = "DEVELOPING"


            model[
                "support_level"
            ] = 1


            model[
                "phase"
            ] = "TRANSFER"


            model[
                "next_action"
            ] = "TRANSFER"


            model[
                "decision_reason"
            ] = (
                "The learner succeeded independently, "
                "so now test the same idea "
                "in a changed situation."
            )


            set_expected(

                model,

                [
                    "TRANSFER",

                    "EXPLANATION"
                ],

                (
                    "The learner independently solves "
                    "a changed-context magnetic-pole "
                    "inference and explains why."
                )
            )


        # --------------------------------------------------------
        # CORRECT DURING TRANSFER
        # --------------------------------------------------------

        elif previous_phase == "TRANSFER":


            model = (
                resolve_misconception_if_ready(
                    model
                )
            )


            model[
                "support_level"
            ] = 1


            model[
                "phase"
            ] = "MASTERY_CHECK"


            model[
                "advance_allowed"
            ] = can_advance(
                model
            )


            if model[
                "advance_allowed"
            ]:


                model[
                    "student_state"
                ] = "SECURE"


                model[
                    "next_action"
                ] = "ADVANCE"


                model[
                    "decision_reason"
                ] = (
                    "The learner now has explanation, "
                    "independent retest, "
                    "and changed-context transfer evidence."
                )


                set_expected(

                    model,

                    [],

                    (
                        "No further evidence is required "
                        "for this learning objective."
                    )
                )


            else:


                model[
                    "student_state"
                ] = "DEVELOPING"


                model[
                    "next_action"
                ] = "PROBE"


                model[
                    "decision_reason"
                ] = (
                    "Transfer succeeded, "
                    "but another required evidence type "
                    "is still missing."
                )


                set_expected(

                    model,

                    [
                        "EXPLANATION"
                    ],

                    (
                        "The learner explains "
                        "the scientific reasoning independently."
                    )
                )


        # --------------------------------------------------------
        # CORRECT AFTER SUPPORT
        # --------------------------------------------------------

        elif (

            previous_action
            in SUPPORTED_ACTIONS

        ):


            if (

                model[
                    "misconception_status"
                ]
                == "ACTIVE"

                and

                "MISCONCEPTION_REVISION"
                in previous_expected_types

            ):


                model = (
                    reduce_misconception(
                        model
                    )
                )


            model[
                "student_state"
            ] = "DEVELOPING"


            # Independent retest means:
            # remove all previous support.

            model[
                "support_level"
            ] = 1


            model[
                "phase"
            ] = "RETEST"


            model[
                "next_action"
            ] = "RETEST"


            model[
                "decision_reason"
            ] = (
                "The learner succeeded with support, "
                "so remove the support "
                "and check the idea independently."
            )


            set_expected(

                model,

                [
                    "INDEPENDENT_RETEST",

                    "EXPLANATION"
                ],

                (
                    "The learner independently answers "
                    "a fresh magnetic-pole case correctly "
                    "without hints or choices."
                )
            )


        # --------------------------------------------------------
        # CORRECT WITHOUT SUPPORT
        # --------------------------------------------------------

        else:


            evidence = model[
                "evidence_matrix"
            ]


            # ----------------------------------------------------
            # NEED EXPLANATION
            # ----------------------------------------------------

            if evidence[
                "explanation"
            ] != "PASSED":


                model[
                    "student_state"
                ] = "DEVELOPING"


                model[
                    "support_level"
                ] = 1


                model[
                    "phase"
                ] = "DIAGNOSIS"


                model[
                    "next_action"
                ] = "PROBE"


                model[
                    "decision_reason"
                ] = (
                    "The answer is correct, "
                    "but the learner still needs "
                    "to show why."
                )


                set_expected(

                    model,

                    [
                        "EXPLANATION"
                    ],

                    (
                        "The learner explains why "
                        "the answer is scientifically correct."
                    )
                )


            # ----------------------------------------------------
            # NEED RETEST
            # ----------------------------------------------------

            elif evidence[
                "independent_retest"
            ] != "PASSED":


                model[
                    "student_state"
                ] = "DEVELOPING"


                model[
                    "support_level"
                ] = 1


                model[
                    "phase"
                ] = "RETEST"


                model[
                    "next_action"
                ] = "RETEST"


                model[
                    "decision_reason"
                ] = (
                    "The learner has explained the idea, "
                    "but it still needs to hold up "
                    "in a fresh independent case."
                )


                set_expected(

                    model,

                    [
                        "INDEPENDENT_RETEST",

                        "EXPLANATION"
                    ],

                    (
                        "The learner independently answers "
                        "a fresh case "
                        "and explains the reasoning."
                    )
                )


            # ----------------------------------------------------
            # NEED TRANSFER
            # ----------------------------------------------------

            elif evidence[
                "transfer"
            ] != "PASSED":


                model[
                    "student_state"
                ] = "DEVELOPING"


                model[
                    "support_level"
                ] = 1


                model[
                    "phase"
                ] = "TRANSFER"


                model[
                    "next_action"
                ] = "TRANSFER"


                model[
                    "decision_reason"
                ] = (
                    "Independent understanding is present, "
                    "so now test the same idea "
                    "in a changed situation."
                )


                set_expected(

                    model,

                    [
                        "TRANSFER",

                        "EXPLANATION"
                    ],

                    (
                        "The learner independently solves "
                        "a changed-context inference "
                        "and explains why."
                    )
                )

    elif classification == "PARTIALLY_CORRECT":


        model[
            "success_count"
        ] = 0


        model[
            "struggle_count"
        ] += 1


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "phase"
        ] = "GUIDED_PRACTICE"


        if model[
            "support_level"
        ] < 2:


            model[
                "support_level"
            ] = 2


            model[
                "next_action"
            ] = "NARROW"


        else:


            model = increase_support(
                model
            )


            level = model[
                "support_level"
            ]


            if level == 3:


                model[
                    "next_action"
                ] = "CONTRAST"


            elif level == 4:


                model[
                    "next_action"
                ] = "HINT"


            elif level == 5:


                model[
                    "next_action"
                ] = "EXAMPLE"


            else:


                model[
                    "next_action"
                ] = "EXPLAIN"


        model[
            "decision_reason"
        ] = (
            "The learner has part of the idea, "
            "so target only the missing "
            "or incorrect part."
        )


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner supplies the missing idea "
                "and gives correct reasoning."
            )
        )


    # ============================================================
    # INSUFFICIENT EVIDENCE
    # ============================================================

    elif classification == "INSUFFICIENT_EVIDENCE":


        model[
            "success_count"
        ] = 0


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "support_level"
        ] = 1


        model[
            "next_action"
        ] = "PROBE"


        model[
            "decision_reason"
        ] = (
            "The learner has not yet given "
            "the specific evidence "
            "the question asked for."
        )


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner states the missing part "
                "of the answer "
                "and explains why."
            )
        )


    # ============================================================
    # INCORRECT
    # ============================================================

    elif classification == "INCORRECT":


        model[
            "success_count"
        ] = 0


        model[
            "struggle_count"
        ] += 1


        model[
            "student_state"
        ] = "NEEDS_REVIEW"


        # --------------------------------------------------------
        # FAILED TRANSFER
        # --------------------------------------------------------

        if previous_phase == "TRANSFER":


            model[
                "phase"
            ] = "DIAGNOSIS"


            model[
                "support_level"
            ] = 2


            model[
                "next_action"
            ] = "PROBE"


            model[
                "decision_reason"
            ] = (
                "The changed-context inference failed, "
                "so diagnose the reasoning "
                "before giving more support."
            )


            set_expected(

                model,

                [
                    "TRANSFER",

                    "EXPLANATION"
                ],

                (
                    "The learner identifies "
                    "the rule needed for the changed situation "
                    "and explains it."
                )
            )


        # --------------------------------------------------------
        # FAILED RETEST
        # --------------------------------------------------------

        elif previous_phase == "RETEST":


            model[
                "phase"
            ] = "DIAGNOSIS"


            model[
                "support_level"
            ] = 2


            model[
                "next_action"
            ] = "DIAGNOSE"


            model[
                "decision_reason"
            ] = (
                "The learner failed an independent retest, "
                "so advancement stays blocked "
                "and the source of the error "
                "must be diagnosed."
            )


            set_expected(

                model,

                [
                    "INDEPENDENT_RETEST",

                    "EXPLANATION"
                ],

                (
                    "The learner reveals the reasoning error "
                    "and later succeeds independently."
                )
            )


        # --------------------------------------------------------
        # NORMAL INCORRECT RESPONSE
        # --------------------------------------------------------

        else:


            model[
                "phase"
            ] = "DIAGNOSIS"


            model[
                "support_level"
            ] = max(

                model[
                    "support_level"
                ],

                2
            )


            model[
                "next_action"
            ] = "DIAGNOSE"


            model[
                "decision_reason"
            ] = (
                "The answer is incorrect, "
                "but there is not yet enough evidence "
                "to name a stable misconception."
            )


            set_expected(

                model,

                [
                    "EXPLANATION"
                ],

                (
                    "The learner reveals the reasoning "
                    "behind the answer."
                )
            )


    # ============================================================
    # MISCONCEPTION
    # ============================================================

    elif classification == "MISCONCEPTION":


        model[
            "success_count"
        ] = 0


        model[
            "struggle_count"
        ] += 1


        model[
            "student_state"
        ] = "NEEDS_REVIEW"


        model = activate_misconception(

            model,

            misconception_id,

            misconception,

            reason
        )


        # --------------------------------------------------------
        # SCIENTIFIC DISTINCTION
        # --------------------------------------------------------

        if model[
            "misconception_id"
        ] in [

            "MAGNETIC_POLES_ARE_ELECTRIC_CHARGES",

            "FIELD_LINES_ARE_PHYSICAL_OBJECTS"

        ]:


            model[
                "phase"
            ] = "GUIDED_PRACTICE"


            model[
                "support_level"
            ] = max(

                model[
                    "support_level"
                ],

                3
            )


            model[
                "next_action"
            ] = "CLARIFY"


            model[
                "decision_reason"
            ] = (
                "Two different scientific ideas "
                "are being mixed together, "
                "so the distinction "
                "must be clarified directly."
            )


            set_expected(

                model,

                [
                    "MISCONCEPTION_REVISION",

                    "EXPLANATION"
                ],

                (
                    "The learner restates "
                    "the scientific distinction correctly "
                    "in their own words."
                )
            )


        # --------------------------------------------------------
        # MISCONCEPTION IN RETEST OR TRANSFER
        # --------------------------------------------------------

        elif previous_phase in [

            "RETEST",

            "TRANSFER"

        ]:


            model[
                "phase"
            ] = "GUIDED_PRACTICE"


            model[
                "support_level"
            ] = 3


            model[
                "next_action"
            ] = "CONTRAST"


            model[
                "decision_reason"
            ] = (
                "The incorrect mental model reappeared "
                "during independent use, "
                "so compare the competing ideas "
                "before another independent try."
            )


            set_expected(

                model,

                [
                    "MISCONCEPTION_REVISION",

                    "EXPLANATION"
                ],

                (
                    "The learner replaces "
                    "the incorrect rule "
                    "with the correct one "
                    "and explains why."
                )
            )


        # --------------------------------------------------------
        # CONTRAST DID NOT WORK
        # --------------------------------------------------------

        elif previous_action == "CONTRAST":


            model[
                "phase"
            ] = "GUIDED_PRACTICE"


            model[
                "support_level"
            ] = 4


            model[
                "next_action"
            ] = "HINT"


            model[
                "decision_reason"
            ] = (
                "The contrast did not change "
                "the mental model, "
                "so give one smaller clue."
            )


            set_expected(

                model,

                [
                    "MISCONCEPTION_REVISION",

                    "EXPLANATION"
                ],

                (
                    "The learner revises "
                    "the incorrect idea "
                    "using the clue "
                    "and explains why."
                )
            )


        # --------------------------------------------------------
        # HINT DID NOT WORK
        # --------------------------------------------------------

        elif previous_action == "HINT":


            model[
                "phase"
            ] = "GUIDED_PRACTICE"


            model[
                "support_level"
            ] = 5


            model[
                "next_action"
            ] = "EXAMPLE"


            model[
                "decision_reason"
            ] = (
                "The hint was not enough, "
                "so use one concrete example."
            )


            set_expected(

                model,

                [
                    "MISCONCEPTION_REVISION",

                    "EXPLANATION"
                ],

                (
                    "The learner uses the example "
                    "to revise the incorrect idea."
                )
            )


        # --------------------------------------------------------
        # FIRST TARGETED MISCONCEPTION RESPONSE
        # --------------------------------------------------------

        else:


            model[
                "phase"
            ] = "GUIDED_PRACTICE"


            model[
                "support_level"
            ] = max(

                model[
                    "support_level"
                ],

                3
            )


            model[
                "next_action"
            ] = "CONTRAST"


            model[
                "decision_reason"
            ] = (
                "The response shows "
                "a specific incorrect mental model, "
                "so compare it directly "
                "with the scientifically correct model."
            )


            set_expected(

                model,

                [
                    "MISCONCEPTION_REVISION",

                    "EXPLANATION"
                ],

                (
                    "The learner revises "
                    "the incorrect model "
                    "and explains the correct distinction."
                )
            )


    # ============================================================
    # VOCABULARY ISSUE
    # ============================================================

    elif classification == "VOCABULARY_ISSUE":


        model[
            "success_count"
        ] = 0


        model[
            "struggle_count"
        ] += 1


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "phase"
        ] = "GUIDED_PRACTICE"


        model[
            "support_level"
        ] = max(

            model[
                "support_level"
            ],

            3
        )


        model[
            "next_action"
        ] = "CLARIFY"


        model[
            "decision_reason"
        ] = (
            "The main barrier "
            "is the meaning of a word, "
            "not yet the science idea itself."
        )


        set_expected(

            model,

            [
                "VOCABULARY"
            ],

            (
                "The learner shows "
                "that the word is understood "
                "and returns to the science question."
            )
        )


    # ============================================================
    # PREREQUISITE GAP
    # ============================================================

    elif classification == "PREREQUISITE_GAP":


        model[
            "success_count"
        ] = 0


        model[
            "struggle_count"
        ] += 1


        model[
            "student_state"
        ] = "NEEDS_REVIEW"


        model[
            "phase"
        ] = "GUIDED_PRACTICE"


        model[
            "support_level"
        ] = max(

            model[
                "support_level"
            ],

            5
        )


        model[
            "next_action"
        ] = "REMEDIATE_PREREQUISITE"


        model[
            "decision_reason"
        ] = (
            "The learner is missing "
            "a foundation needed "
            "to make sense "
            "of the current idea."
        )


        set_expected(

            model,

            [
                "PREREQUISITE"
            ],

            (
                "The learner shows "
                "the minimum prerequisite idea "
                "before returning "
                "to the original objective."
            )
        )


    # ============================================================
    # UNCERTAIN
    # ============================================================

    elif classification == "UNCERTAIN":


        model[
            "success_count"
        ] = 0


        model[
            "struggle_count"
        ] += 1


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "phase"
        ] = "DIAGNOSIS"


        struggle = model[
            "struggle_count"
        ]


        if struggle == 1:


            model[
                "support_level"
            ] = 1


            model[
                "next_action"
            ] = "PROBE"


            model[
                "decision_reason"
            ] = (
                "The learner is uncertain, "
                "so invite one small guess "
                "before adding support."
            )


        elif struggle == 2:


            model[
                "support_level"
            ] = 2


            model[
                "next_action"
            ] = "NARROW"


            model[
                "decision_reason"
            ] = (
                "The learner is still uncertain, "
                "so make the question smaller."
            )


        elif struggle == 3:


            model[
                "support_level"
            ] = 4


            model[
                "next_action"
            ] = "HINT"


            model[
                "decision_reason"
            ] = (
                "The learner has tried "
                "without success, "
                "so give one small clue."
            )


        elif struggle == 4:


            model[
                "support_level"
            ] = 5


            model[
                "next_action"
            ] = "EXAMPLE"


            model[
                "decision_reason"
            ] = (
                "The clue was not enough, "
                "so use one concrete example."
            )


        else:


            model[
                "support_level"
            ] = 6


            model[
                "next_action"
            ] = "EXPLAIN"


            model[
                "decision_reason"
            ] = (
                "Repeated uncertainty "
                "now justifies "
                "a short direct explanation "
                "before trying again."
            )


        set_expected(

            model,

            [
                "PREDICTION",

                "PRIOR_KNOWLEDGE"
            ],

            (
                "The learner makes "
                "an initial guess "
                "or reveals "
                "what part is confusing."
            )
        )


    # ============================================================
    # IRRELEVANT
    # ============================================================

    elif classification == "IRRELEVANT":


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        model[
            "decision_reason"
        ] = (
            "The response does not provide evidence "
            "for the current idea, "
            "so gently reconnect it."
        )


        set_expected(

            model,

            [
                "PRIOR_KNOWLEDGE",

                "PREDICTION"
            ],

            (
                "The learner reconnects "
                "to the current science idea."
            )
        )


    # ============================================================
    # UNKNOWN
    # ============================================================

    else:


        model[
            "student_state"
        ] = "DEVELOPING"


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "support_level"
        ] = max(

            model[
                "support_level"
            ],

            2
        )


        model[
            "next_action"
        ] = "DIAGNOSE"


        model[
            "decision_reason"
        ] = (
            "The classifier result is unclear, "
            "so collect one more piece of evidence."
        )


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner gives enough information "
                "for a clearer diagnosis."
            )
        )


    # ============================================================
    # SEMANTIC EVENT
    # ============================================================

    model[
        "last_semantic_event"
    ] = derive_semantic_event(

        classification,

        previous_action,

        previous_phase,

        model
    )


    # ============================================================
    # SYNCHRONISE FLAGS
    # ============================================================

    model[
        "retest_passed"
    ] = (

        model[
            "evidence_matrix"
        ][
            "independent_retest"
        ]
        == "PASSED"
    )


    model[
        "transfer_passed"
    ] = (

        model[
            "evidence_matrix"
        ][
            "transfer"
        ]
        == "PASSED"
    )


    # ============================================================
    # ADVANCEMENT GATE
    # ============================================================

    model[
        "advance_allowed"
    ] = can_advance(
        model
    )


    if (

        model[
            "next_action"
        ] == "ADVANCE"

        and

        not model[
            "advance_allowed"
        ]

    ):


        model[
            "student_state"
        ] = "DEVELOPING"


        evidence = model[
            "evidence_matrix"
        ]


        if evidence[
            "explanation"
        ] != "PASSED":


            model[
                "phase"
            ] = "DIAGNOSIS"


            model[
                "next_action"
            ] = "PROBE"


            set_expected(

                model,

                [
                    "EXPLANATION"
                ],

                (
                    "The learner explains "
                    "the scientific reasoning independently."
                )
            )


        elif evidence[
            "independent_retest"
        ] != "PASSED":


            model[
                "phase"
            ] = "RETEST"


            model[
                "next_action"
            ] = "RETEST"


            set_expected(

                model,

                [
                    "INDEPENDENT_RETEST",

                    "EXPLANATION"
                ],

                (
                    "The learner independently answers "
                    "a fresh case without support."
                )
            )


        else:


            model[
                "phase"
            ] = "TRANSFER"


            model[
                "next_action"
            ] = "TRANSFER"


            set_expected(

                model,

                [
                    "TRANSFER",

                    "EXPLANATION"
                ],

                (
                    "The learner independently solves "
                    "a changed-context inference "
                    "and explains why."
                )
            )


        model[
            "decision_reason"
        ] = (
            "Advancement is blocked "
            "because required evidence "
            "is still missing."
        )


    return model


# ================================================================
# STUDENT-FACING TUTOR
# ================================================================

def generate_tutor_response(

    model,

    classification_result,

    student_message,

    history_string
):


    prompt = f"""

You are the student-facing voice
of a Class 6 Science Tutor.


The Python Tutor Engine
has already selected
the teaching action.


Follow that action exactly.


CURRENT CONCEPT:

{model["concept"]}


LEARNING OBJECTIVE:

{learning_objective}


STUDENT STATE:

{model["student_state"]}


PHASE:

{model["phase"]}


SUPPORT LEVEL:

{model["support_level"]}


MISCONCEPTION:

{model["misconception"]}


MISCONCEPTION STATUS:

{model["misconception_status"]}


SELECTED ACTION:

{model["next_action"]}


SEMANTIC EVENT:

{model["last_semantic_event"]}


EXPECTED NEXT EVIDENCE:

{model["expected_next_evidence"]}


ADVANCE ALLOWED:

{model["advance_allowed"]}


CLASSIFIER RESULT:

{classification_result}


CURRENT STUDENT MESSAGE:

{student_message}


RECENT CONVERSATION:

{history_string}


================================================
MOST IMPORTANT RULE
================================================

Keep the TEACHING LOGIC
exactly as selected
by the Tutor Engine.


Only make the LANGUAGE
simple,
natural,
and comfortable
for a Class 6 student.


Do NOT change the selected action
just because another response
sounds easier.


================================================
HOW IT SHOULD FEEL
================================================

The learner should feel like:


"I figured something out."


not:


"I read the answer from a textbook."


The Tutor should feel like
a very good science guide.


Curious.

Clear.

Patient.

Smart.

Natural.


Not robotic.


Not childish.


Not formal.


================================================
LANGUAGE
================================================

Speak naturally
to an 11-12 year old.


Use short,
clear sentences.


Use normal everyday words
where possible.


If a scientific word
is important,
use the real scientific word,

but explain it simply.


Do NOT sound like:


a textbook,

an exam paper,

a research paper,

a dictionary,

or a formal lecture.


Do NOT baby-talk.


Do NOT expose internal words
such as:


classification

evidence matrix

phase

student state

support level

misconception status

mastery

advance allowed


Usually use about
25-85 words.


Ask only ONE main question
at a time.


================================================
VERY IMPORTANT FIRST-TURN RULE
================================================

If SELECTED ACTION is PROBE

and the learner asks something like:


"What is a magnet?"


"What are magnetic poles?"


"What happens with magnets?"


DO NOT immediately give
the full definition or answer.


First discover
what the learner already thinks.


Example:


Student:

"What is a magnet?"


Tutor:

"What do you think
a magnet is or does?

Even a small guess is okay."


This is especially important
at the beginning
of the conversation.


================================================
DO NOT OVER-PRAISE
================================================

Do not automatically say:


Amazing!

Perfect!

Brilliant!

Great job!


Use specific feedback instead.


Example:


Student:

"I think magnets repel."


Better:


"You noticed the pushing part."


Then continue
with the next useful question.


================================================
PROBE
================================================

If action is PROBE:


Do NOT immediately teach
the answer.


Ask for ONE missing thought.


Find out what
the learner already thinks.


Examples:


"What do you think?"


"What made you choose that?"


"What do you think
a magnet can do?"


"Which part
are you most sure about?"


One useful question
is enough.


================================================
NARROW
================================================

If action is NARROW:


Use what the learner
just said.


Make the next question
a little smaller.


Do NOT suddenly explain everything.


Example:


Student:

"I think a magnet
is something that repels."


Tutor:

"You noticed
the pushing part.

Do you think magnets
always push away,

or can they sometimes
pull together too?"


The goal is to uncover
the learner's next thought.


================================================
CONTRAST
================================================

If action is CONTRAST:


Put TWO competing ideas
side by side.


The contrast must directly target
the learner's real misconception.


Example:


If learner believes:


"North is positive
and south is negative."


You could say:


"Two ideas are getting mixed here:


A. North and south are
electric + and - charges.


B. North and south are
names of magnetic poles.


Which one do you think
fits magnets better?"


Ask ONE reason.


Do not immediately reveal
the answer
unless stronger support
is required.


================================================
HINT
================================================

If action is HINT:


Give ONE small clue.


Do NOT reveal
the complete answer.


================================================
EXAMPLE
================================================

If action is EXAMPLE:


Give ONE simple,
concrete example.


If it is an analogy,
make that clear
when necessary.


Do not create
a false scientific model.


================================================
EXPLAIN
================================================

If action is EXPLAIN:


Give only
the minimum explanation needed.


Do not lecture.


Use something like:


simple idea

->

correct science word

->

tiny example

->

fresh try


================================================
CLARIFY
================================================

If action is CLARIFY:


Correct ONE precise
scientific distinction
in simple English.


Then ask ONE
small check question.


For magnetic poles
versus electrical charge:


North and south
are magnetic pole names.


They are NOT
positive and negative
electric charges.


For magnetic field lines:


Magnetic field lines
are a way scientists
show or represent
the magnetic field.


They are NOT
physical strings,
wires,
ropes,
or objects
that push magnets.


Magnets exert
magnetic forces
on each other.


================================================
REMEDIATE_PREREQUISITE
================================================

Use this action only when
the learner has clearly shown
that a basic idea is missing.


Do NOT use it merely because
the learner asked:


"What is a magnet?"


If the learner says:


"I don't know
what a magnet is."


or


"I don't understand
what a magnetic pole is."


then teach only
the missing foundation.


For this objective:


1. A magnet can pull
   or push another magnet.


2. A magnet has two poles
   called north and south.


3. North and south
   are magnetic pole names.


4. They are NOT
   electric + and - charges.


Then ask ONE tiny check.


================================================
RETEST
================================================

If action is RETEST:


Remove previous support.


No hint.


No multiple choice.


Do not hide the answer
inside the question.


Ask one fresh case.


Example:


"Try this one
without a clue:

what happens
when two south poles
face each other?

What tells you that?"


================================================
TRANSFER
================================================

If action is TRANSFER:


Use the SAME concept
in a changed situation.


The learner must infer
something unknown.


Example:


"Two magnets
push away from each other.


One facing pole is North.


What must
the other facing pole be?


How do you know?"


Do NOT reveal
the answer
inside the question.


Do NOT simply ask
the learner to repeat:


"Like poles repel
and unlike poles attract."


Do NOT suddenly move to:


paperclips,

electromagnets,

magnetic materials,

compasses,

or another concept.


Transfer means:


same idea

+

new situation


================================================
ADVANCE
================================================

If action is ADVANCE:


Move forward only when:


ADVANCE ALLOWED = True


Never override
the engine.


================================================
CONTRADICTION HANDLING
================================================

Compare the current message
with RECENT CONVERSATION.


If the learner now says
something that conflicts
with an earlier answer,


do NOT silently overwrite it.


Point it out gently.


Example:


"Earlier you said
like poles repel,

but now you're saying
two north poles attract.


Which idea
do you want to keep,

and why?"


================================================
PREDICTION VS OBSERVATION
================================================

Prediction:


what the learner thinks
will happen.


Observation:


what the learner actually
reports seeing.


Do not pretend
an experiment happened
if the learner
did not report it.


================================================
REAL-WORLD CONNECTIONS
================================================

You may connect science
to familiar things
when useful.


But do NOT let
the surroundings
become a random new topic.


Do not assume
the learner owns magnets.


If suggesting an observation,
make it optional.


================================================
SCIENCE GUARDRAILS
================================================

Never say:


"Magnets want
to come together."


Never say:


"All metals
are magnetic."


Never say:


"Field lines physically
push magnets."


Never say:


"North is positive
and south is negative."


Use simple language,

but keep the science correct.


================================================
SAFETY
================================================

Never ask a child to:


touch sockets,

open electrical appliances,

use mains electricity,

mix unknown chemicals,

look directly at the Sun,

handle flames alone,

touch dangerously hot objects,

or climb unsafe places.


================================================
IDEAL FLOW
================================================

learner's idea

->

small question

->

notice a pattern

->

better question

->

explain

->

try independently

->

use the idea
in a new situation


Do not make it feel
like memorising
a school book.

"""


    response = client.chat.completions.create(

        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],

        model=MODEL_NAME,

        temperature=0.3
    )


    return response.choices[0].message.content


# ================================================================
# DEBUG PANEL
# ================================================================

def print_debug(

    model,

    student_response
):


    print(
        "\n"
        + "=" * 70
    )


    print(
        "V3.1 TUTOR DEBUG"
    )


    print(
        "=" * 70
    )


    print(
        "CURRENT CONCEPT:",
        model[
            "concept"
        ]
    )


    print(
        "LEARNING OBJECTIVE:",
        learning_objective
    )


    print(
        "STUDENT RESPONSE:",
        student_response
    )


    print(
        "CLASSIFICATION:",
        model[
            "last_classification"
        ]
    )


    print(
        "SEMANTIC EVENT:",
        model[
            "last_semantic_event"
        ]
    )


    print(
        "STUDENT STATE:",
        model[
            "student_state"
        ]
    )


    print(
        "PHASE:",
        model[
            "phase"
        ]
    )


    print(
        "SUPPORT LEVEL:",
        model[
            "support_level"
        ]
    )


    print(
        "DIFFICULTY:",
        model[
            "difficulty"
        ]
    )


    print(
        "MISCONCEPTION ID:",
        model[
            "misconception_id"
        ]
    )


    print(
        "MISCONCEPTION:",
        model[
            "misconception"
        ]
    )


    print(
        "MISCONCEPTION STATUS:",
        model[
            "misconception_status"
        ]
    )


    print(
        "NEXT ACTION:",
        model[
            "next_action"
        ]
    )


    print(
        "DECISION REASON:",
        model[
            "decision_reason"
        ]
    )


    print(
        "EXPECTED EVIDENCE TYPES:",
        model[
            "expected_evidence_types"
        ]
    )


    print(
        "EXPECTED NEXT EVIDENCE:",
        model[
            "expected_next_evidence"
        ]
    )


    print(
        "RETEST PASSED:",
        model[
            "retest_passed"
        ]
    )


    print(
        "TRANSFER PASSED:",
        model[
            "transfer_passed"
        ]
    )


    print(
        "\nEVIDENCE MATRIX:"
    )


    for key, value in (
        model[
            "evidence_matrix"
        ].items()
    ):


        print(
            f"  {key}: {value}"
        )


    print(
        "\nADVANCE ALLOWED:",
        model[
            "advance_allowed"
        ]
    )


    print(
        "=" * 70
    )


# ================================================================
# NORMAL TUTOR
# ================================================================

def run_tutor():


    global student_model


    student_model = (
        create_student_model()
    )


    history_turns = []


    memory_limit = 20


    last_tutor_response = ""


    print(
        "\n"
        + "=" * 70
    )


    print(
        "             CLASS 6 SCIENCE AI TUTOR"
    )


    print(
        "=" * 70
    )


    print(
        """

Hello!!

Welcome to your Class 6 Science AI Tutor.

I will connect Science with things around you,
help you observe carefully,
make predictions,
test ideas safely,
and gradually discover deeper explanations.

Science is not only about finding answers.
It is also about learning how to ask
better questions.

Type 'exit' to stop.

"""
    )


    while True:


        question = input(
            "\nStudent: "
        )


        if (
            question
            .strip()
            .lower()
            == "exit"
        ):


            print(
                "\nTutor:"
            )


            print(
                "Goodbye! Keep observing, "
                "wondering, and asking better questions."
            )


            break


        if not question.strip():


            continue


        history_string = "\n\n".join(
            history_turns
        )

        if last_tutor_response:


            classification_result = (
                classify_student_response(

                    current_concept=
                        current_concept,

                    learning_objective=
                        learning_objective,

                    tutor_question=
                        last_tutor_response,

                    student_response=
                        question,

                    history_string=
                        history_string,

                    expected_evidence_types=
                        student_model[
                            "expected_evidence_types"
                        ],

                    expected_next_evidence=
                        student_model[
                            "expected_next_evidence"
                        ]
                )
            )


        else:


            classification_result = (

                "CLASSIFICATION: INITIAL_QUERY\n"

                "REASON: "
                "This is the first student message.\n"

                "MISCONCEPTION_ID: NONE\n"

                "MISCONCEPTION: NONE"
            )


        # ========================================================
        # UPDATE STUDENT MODEL
        # ========================================================

        student_model = (
            update_tutor_state(

                classification_result,

                student_model
            )
        )


        # ========================================================
        # DEBUG
        # ========================================================

        print_debug(

            student_model,

            question
        )


        # ========================================================
        # TUTOR RESPONSE
        # ========================================================

        tutor_response = (
            generate_tutor_response(

                student_model,

                classification_result,

                question,

                history_string
            )
        )


        print(
            "\nTutor:"
        )


        print(
            tutor_response
        )


        # ========================================================
        # MEMORY
        # ========================================================

        history_turns.append(

            "Student:\n"
            + question
            + "\n\nTutor:\n"
            + tutor_response
        )


        if len(
            history_turns
        ) > memory_limit:


            history_turns.pop(0)


        last_tutor_response = (
            tutor_response
        )


def prepare_validation_state(
    test_id
):


    model = create_student_model()


    model[
        "student_state"
    ] = "DEVELOPING"

    if test_id == "T01":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "PREDICTION"
            ],

            (
                "The learner makes a prediction "
                "about what happens "
                "when two north poles face each other."
            )
        )


    # ------------------------------------------------------------
    # T02
    # ------------------------------------------------------------

    elif test_id == "T02":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "PREDICTION",

                "EXPLANATION"
            ],

            (
                "The learner predicts the interaction "
                "and explains the reasoning."
            )
        )


    # ------------------------------------------------------------
    # T03
    # ------------------------------------------------------------

    elif test_id == "T03":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "PREDICTION",

                "PRIOR_KNOWLEDGE"
            ],

            (
                "The learner makes an initial prediction "
                "or reveals relevant prior knowledge."
            )
        )


    # ------------------------------------------------------------
    # T04
    # ------------------------------------------------------------

    elif test_id == "T04":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner explains "
                "what like poles mean "
                "and gives correct reasoning."
            )
        )


    # ------------------------------------------------------------
    # T05
    # ------------------------------------------------------------

    elif test_id == "T05":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner states "
                "which pole is unknown "
                "and explains why."
            )
        )


    # ------------------------------------------------------------
    # T06
    # ------------------------------------------------------------

    elif test_id == "T06":


        model[
            "phase"
        ] = "GUIDED_PRACTICE"


        model[
            "support_level"
        ] = 4


        model[
            "next_action"
        ] = "HINT"


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner explains "
                "the correct interaction "
                "using scientifically correct reasoning."
            )
        )


    # ------------------------------------------------------------
    # T07
    # ------------------------------------------------------------

    elif test_id == "T07":


        model[
            "student_state"
        ] = "NEEDS_REVIEW"


        model[
            "phase"
        ] = "GUIDED_PRACTICE"


        model[
            "support_level"
        ] = 4


        model[
            "next_action"
        ] = "HINT"


        model[
            "misconception_id"
        ] = "LIKE_POLES_ATTRACT"


        model[
            "misconception"
        ] = (
            "The learner believes "
            "that like magnetic poles attract."
        )


        model[
            "misconception_status"
        ] = "ACTIVE"


        set_expected(

            model,

            [
                "MISCONCEPTION_REVISION",

                "EXPLANATION"
            ],

            (
                "The learner revises "
                "the incorrect idea "
                "using scientifically correct reasoning."
            )
        )


    # ------------------------------------------------------------
    # T08
    # ------------------------------------------------------------

    elif test_id == "T08":


        model[
            "phase"
        ] = "RETEST"


        model[
            "support_level"
        ] = 1


        model[
            "next_action"
        ] = "RETEST"


        model[
            "misconception_id"
        ] = "LIKE_POLES_ATTRACT"


        model[
            "misconception"
        ] = (
            "The learner previously believed "
            "that like poles attract."
        )


        model[
            "misconception_status"
        ] = "REDUCED"


        model[
            "evidence_matrix"
        ][
            "prior_knowledge"
        ] = "COLLECTED"


        model[
            "evidence_matrix"
        ][
            "explanation"
        ] = "PASSED"


        model[
            "evidence_matrix"
        ][
            "guided_application"
        ] = "PASSED"


        set_expected(

            model,

            [
                "INDEPENDENT_RETEST",

                "EXPLANATION"
            ],

            (
                "The learner independently answers "
                "a fresh magnetic-pole case correctly "
                "without support."
            )
        )


    # ------------------------------------------------------------
    # T09
    # ------------------------------------------------------------

    elif test_id == "T09":


        model[
            "phase"
        ] = "RETEST"


        model[
            "support_level"
        ] = 1


        model[
            "next_action"
        ] = "RETEST"


        model[
            "evidence_matrix"
        ][
            "prior_knowledge"
        ] = "COLLECTED"


        model[
            "evidence_matrix"
        ][
            "prediction"
        ] = "PASSED"


        model[
            "evidence_matrix"
        ][
            "explanation"
        ] = "PASSED"


        model[
            "evidence_matrix"
        ][
            "guided_application"
        ] = "PASSED"


        set_expected(

            model,

            [
                "INDEPENDENT_RETEST",

                "EXPLANATION"
            ],

            (
                "The learner independently answers "
                "a fresh magnetic-pole case correctly."
            )
        )


    # ------------------------------------------------------------
    # T10 / T11 / T14 / T15
    # ------------------------------------------------------------

    elif test_id in [

        "T10",

        "T11",

        "T14",

        "T15"

    ]:


        model[
            "phase"
        ] = "TRANSFER"


        model[
            "support_level"
        ] = 1


        model[
            "next_action"
        ] = "TRANSFER"


        model[
            "evidence_matrix"
        ][
            "prior_knowledge"
        ] = "COLLECTED"


        model[
            "evidence_matrix"
        ][
            "prediction"
        ] = "PASSED"


        model[
            "evidence_matrix"
        ][
            "explanation"
        ] = "PASSED"


        model[
            "evidence_matrix"
        ][
            "guided_application"
        ] = "PASSED"


        model[
            "evidence_matrix"
        ][
            "independent_retest"
        ] = "PASSED"


        model[
            "retest_passed"
        ] = True


        set_expected(

            model,

            [
                "TRANSFER",

                "EXPLANATION"
            ],

            (
                "The learner independently solves "
                "a changed-context magnetic-pole inference "
                "and explains why."
            )
        )


        if test_id == "T11":


            model[
                "misconception_id"
            ] = "LIKE_POLES_ATTRACT"


            model[
                "misconception"
            ] = (
                "The learner previously believed "
                "that like poles attract."
            )


            model[
                "misconception_status"
            ] = "REDUCED"


        if test_id == "T14":


            model[
                "misconception_id"
            ] = "TRANSFER_RULE_REVERSAL"


            model[
                "misconception"
            ] = (
                "The learner previously confused "
                "which pole relationships repel."
            )


            model[
                "misconception_status"
            ] = "REDUCED"


    # ------------------------------------------------------------
    # T12
    # ------------------------------------------------------------

    elif test_id == "T12":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner explains whether "
                "north and south magnetic poles "
                "are positive and negative "
                "electrical charges."
            )
        )


    # ------------------------------------------------------------
    # T13
    # ------------------------------------------------------------

    elif test_id == "T13":


        model[
            "phase"
        ] = "DIAGNOSIS"


        model[
            "next_action"
        ] = "PROBE"


        set_expected(

            model,

            [
                "EXPLANATION"
            ],

            (
                "The learner explains "
                "what magnetic field lines represent "
                "and whether they are physical objects."
            )
        )


    return model


# ================================================================
# FIXED T01-T15 CASES
# ================================================================

VALIDATION_CASES = {


    "T01": {

        "question":
            (
                "What happens when two north poles "
                "are brought close together?"
            ),

        "response":
            "North and north repel.",

        "classification":
            [
                "CORRECT"
            ],

        "state":
            [
                "DEVELOPING"
            ],

        "misconception_status":
            [
                None
            ],

        "action":
            [
                "PROBE"
            ],

        "advance":
            False,

        "evidence": {

            "prediction":
                "PASSED"
        }
    },


    "T02": {

        "question":
            (
                "Two north poles face each other. "
                "What happens, and why?"
            ),

        "response":
            (
                "North and north attract "
                "because both are magnets."
            ),

        "classification":
            [
                "MISCONCEPTION"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE"
            ],

        "action":
            [
                "CONTRAST"
            ],

        "advance":
            False,

        "evidence":
            {}
    },


    "T03": {

        "question":
            (
                "What do you think happens "
                "when two north poles face each other?"
            ),

        "response":
            "I don't know.",

        "classification":
            [
                "UNCERTAIN"
            ],

        "state":
            [
                "DEVELOPING"
            ],

        "misconception_status":
            [
                None
            ],

        "action":
            [
                "PROBE",

                "HINT"
            ],

        "advance":
            False,

        "evidence": {

            "prior_knowledge":
                "PENDING"
        }
    },


    "T04": {

        "question":
            (
                "Like poles repel. "
                "What does 'like' mean here, "
                "and why?"
            ),

        "response":
            (
                "Like poles repel "
                "because they are different."
            ),

        "classification":
            [
                "MISCONCEPTION",

                "PARTIALLY_CORRECT"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE"
            ],

        "action":
            [
                "CONTRAST"
            ],

        "advance":
            False,

        "evidence":
            {}
    },


    "T05": {

        "question":
            (
                "An unknown magnetic pole attracts "
                "a north pole. "
                "What is the unknown pole, "
                "and why?"
            ),

        "response":
            "They attract.",

        "classification":
            [
                "INSUFFICIENT_EVIDENCE",

                "PARTIALLY_CORRECT"
            ],

        "state":
            [
                "DEVELOPING"
            ],

        "misconception_status":
            [
                None
            ],

        "action":
            [
                "PROBE"
            ],

        "advance":
            False,

        "evidence":
            {}
    },


    "T06": {

        "question":
            (
                "Using the hint, "
                "what happens when north and south "
                "face each other, and why?"
            ),

        "response":
            (
                "North and south attract "
                "because unlike poles attract."
            ),

        "classification":
            [
                "CORRECT"
            ],

        "semantic_event":
            [
                "CORRECT_AFTER_SUPPORT"
            ],

        "state":
            [
                "DEVELOPING"
            ],

        "misconception_status":
            [
                None
            ],

        "action":
            [
                "RETEST"
            ],

        "advance":
            False,

        "evidence": {

            "explanation":
                "PASSED",

            "guided_application":
                "PASSED",

            "independent_retest":
                "PENDING"
        }
    },


    "T07": {

        "question":
            (
                "Using the hint, "
                "what happens when north faces north, "
                "and why?"
            ),

        "response":
            (
                "North and north repel "
                "because like poles repel."
            ),

        "classification":
            [
                "CORRECT"
            ],

        "semantic_event":
            [
                "CORRECT_AFTER_SUPPORT"
            ],

        "state":
            [
                "DEVELOPING"
            ],

        "misconception_status":
            [
                "REDUCED"
            ],

        "action":
            [
                "RETEST"
            ],

        "advance":
            False,

        "evidence": {

            "guided_application":
                "PASSED",

            "independent_retest":
                "PENDING"
        }
    },


    "T08": {

        "question":
            (
                "Without any hint, "
                "what happens when two north poles "
                "face each other?"
            ),

        "response":
            (
                "North and north attract."
            ),

        "classification":
            [
                "INCORRECT",

                "MISCONCEPTION"
            ],

        "semantic_event":
            [
                "FAILED_RETEST"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE",

                "REDUCED"
            ],

        "action":
            [
                "DIAGNOSE",

                "CONTRAST"
            ],

        "advance":
            False,

        "evidence": {

            "independent_retest":
                "FAILED"
        }
    },


    "T09": {

        "question":
            (
                "Without a hint, "
                "compare north-north and north-south. "
                "What happens in each case, and why?"
            ),

        "response":
            (
                "North and north repel, "
                "and north and south attract, "
                "because like poles repel "
                "and unlike poles attract."
            ),

        "classification":
            [
                "CORRECT"
            ],

        "semantic_event":
            [
                "RETEST_PASSED"
            ],

        "state":
            [
                "DEVELOPING"
            ],

        "misconception_status":
            [
                None,

                "REDUCED"
            ],

        "action":
            [
                "TRANSFER"
            ],

        "advance":
            False,

        "evidence": {

            "independent_retest":
                "PASSED",

            "transfer":
                "PENDING"
        }
    },


    "T10": {

        "question":
            (
                "Magnet A is North. "
                "Magnet B pushes away from A. "
                "What must B be? "
                "Explain why."
            ),

        "response":
            (
                "B must be South "
                "because north and south attract."
            ),

        "classification":
            [
                "MISCONCEPTION"
            ],

        "semantic_event":
            [
                "FAILED_TRANSFER"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE"
            ],

        "action":
            [
                "CONTRAST",

                "PROBE"
            ],

        "advance":
            False,

        "evidence": {

            "independent_retest":
                "PASSED",

            "transfer":
                "FAILED"
        }
    },


    "T11": {

        "question":
            (
                "Magnet A is North. "
                "Magnet B pushes away from A. "
                "What must B be? "
                "Explain why."
            ),

        "response":
            (
                "B must be North "
                "because like poles repel."
            ),

        "classification":
            [
                "CORRECT"
            ],

        "semantic_event":
            [
                "TRANSFER_PASSED"
            ],

        "state":
            [
                "SECURE",

                "DEVELOPING"
            ],

        "misconception_status":
            [
                "RESOLVED",

                None
            ],

        "action":
            [
                "ADVANCE",

                "PROBE"
            ],

        "advance":
            True,

        "evidence": {

            "independent_retest":
                "PASSED",

            "transfer":
                "PASSED"
        }
    },


    "T12": {

        "question":
            (
                "Are north and south magnetic poles "
                "the same thing as positive "
                "and negative electrical charges?"
            ),

        "response":
            (
                "North is positive "
                "and south is negative."
            ),

        "classification":
            [
                "MISCONCEPTION"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE"
            ],

        "action":
            [
                "CLARIFY",

                "CONTRAST"
            ],

        "advance":
            False,

        "evidence":
            {}
    },


    "T13": {

        "question":
            (
                "What do magnetic field lines represent? "
                "Are they physical things?"
            ),

        "response":
            (
                "Field lines physically push "
                "the magnets apart."
            ),

        "classification":
            [
                "MISCONCEPTION"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE"
            ],

        "action":
            [
                "CLARIFY"
            ],

        "advance":
            False,

        "evidence":
            {}
    },


    "T14": {

        "question":
            (
                "After earlier help, "
                "here is a new case: "
                "A is North and the magnets repel. "
                "What must B be, and why?"
            ),

        "response":
            (
                "B must be South "
                "because north and south attract."
            ),

        "classification":
            [
                "INCORRECT",

                "MISCONCEPTION"
            ],

        "semantic_event":
            [
                "FAILED_TRANSFER"
            ],

        "state":
            [
                "NEEDS_REVIEW"
            ],

        "misconception_status":
            [
                "ACTIVE",

                "REDUCED"
            ],

        "action":
            [
                "CONTRAST",

                "PROBE"
            ],

        "advance":
            False,

        "evidence": {

            "independent_retest":
                "PASSED",

            "transfer":
                "FAILED"
        }
    },


    "T15": {

        "question":
            (
                "A's facing pole is South. "
                "The magnets attract. "
                "What must B's facing pole be? "
                "Explain why."
            ),

        "response":
            (
                "B must be North "
                "because unlike poles attract."
            ),

        "classification":
            [
                "CORRECT"
            ],

        "semantic_event":
            [
                "TRANSFER_PASSED"
            ],

        "state":
            [
                "SECURE"
            ],

        "misconception_status":
            [
                None,

                "RESOLVED"
            ],

        "action":
            [
                "ADVANCE"
            ],

        "advance":
            True,

        "evidence": {

            "independent_retest":
                "PASSED",

            "transfer":
                "PASSED"
        }
    }
}

def expected_evidence_controls_action(
    model
):


    types = model[
        "expected_evidence_types"
    ]


    action = model[
        "next_action"
    ]


    if action == "RETEST":


        return (

            "INDEPENDENT_RETEST"
            in types
        )


    if action == "TRANSFER":


        return (

            "TRANSFER"
            in types
        )


    if action == "ADVANCE":


        return (

            model[
                "advance_allowed"
            ]

            and

            len(types) == 0
        )


    if (

        action
        in [
            "CONTRAST",

            "CLARIFY"
        ]

        and

        model[
            "misconception_status"
        ] == "ACTIVE"

    ):


        return (

            "MISCONCEPTION_REVISION"
            in types
        )


    if action == "REMEDIATE_PREREQUISITE":


        return (

            "PREREQUISITE"
            in types
        )


    return (

        len(types) > 0
    )


def decision_reason_matches_state(
    model
):


    reason = (

        model[
            "decision_reason"
        ]

        or ""

    ).lower()


    action = model[
        "next_action"
    ]


    classification = model[
        "last_classification"
    ]


    if not reason.strip():


        return False


    if action == "RETEST":


        return (

            "independent"
            in reason

            or

            "support"
            in reason
        )


    if action == "TRANSFER":


        return (

            "changed"
            in reason

            or

            "transfer"
            in reason
        )


    if action == "ADVANCE":


        return (

            "evidence"
            in reason

            or

            "explanation"
            in reason
        )


    if classification == "INSUFFICIENT_EVIDENCE":


        return (

            "evidence"
            in reason
        )


    if classification == "UNCERTAIN":


        return (

            "uncertain"
            in reason
        )


    if classification == "PREREQUISITE_GAP":


        return (

            "foundation"
            in reason
        )


    if classification == "VOCABULARY_ISSUE":


        return (

            "word"
            in reason
        )


    if model[
        "misconception_status"
    ] == "ACTIVE":


        return any(

            word
            in reason

            for word in [

                "incorrect",

                "mental model",

                "scientific ideas"
            ]
        )


    return True


# ================================================================
# RUN ONE VALIDATION CASE
# ================================================================

def run_validation_case(

    test_id,

    show=True
):


    case = VALIDATION_CASES[
        test_id
    ]


    model = prepare_validation_state(
        test_id
    )


    classification_result = (
        classify_student_response(

            current_concept,

            learning_objective,

            case[
                "question"
            ],

            case[
                "response"
            ],

            "",

            model[
                "expected_evidence_types"
            ],

            model[
                "expected_next_evidence"
            ]
        )
    )


    model = update_tutor_state(

        classification_result,

        model
    )


    checks = {


        "classification":

            model[
                "last_classification"
            ]

            in case[
                "classification"
            ],


        "state":

            model[
                "student_state"
            ]

            in case[
                "state"
            ],


        "misconception_status":

            model[
                "misconception_status"
            ]

            in case[
                "misconception_status"
            ],


        "action":

            model[
                "next_action"
            ]

            in case[
                "action"
            ],


        "advance":

            model[
                "advance_allowed"
            ]

            == case[
                "advance"
            ],


        "expected_evidence_control":

            expected_evidence_controls_action(
                model
            ),


        "decision_reason":

            decision_reason_matches_state(
                model
            )
    }


    if (
        "semantic_event"
        in case
    ):


        checks[
            "semantic_event"
        ] = (

            model[
                "last_semantic_event"
            ]

            in case[
                "semantic_event"
            ]
        )


    for key, expected in (
        case[
            "evidence"
        ].items()
    ):


        checks[
            key
        ] = (

            model[
                "evidence_matrix"
            ][key]

            == expected
        )


    passed = all(
        checks.values()
    )


    result = {

        "id":
            test_id,

        "passed":
            passed,

        "checks":
            checks,

        "model":
            model
    }


    if show:


        print(
            "\n"
            + "=" * 72
        )


        print(
            test_id,
            "VALIDATION"
        )


        print(
            "=" * 72
        )


        print(
            "CLASSIFICATION:",
            model[
                "last_classification"
            ]
        )


        print(
            "SEMANTIC EVENT:",
            model[
                "last_semantic_event"
            ]
        )


        print(
            "STATE:",
            model[
                "student_state"
            ]
        )


        print(
            "PHASE:",
            model[
                "phase"
            ]
        )


        print(
            "SUPPORT LEVEL:",
            model[
                "support_level"
            ]
        )


        print(
            "MISCONCEPTION ID:",
            model[
                "misconception_id"
            ]
        )


        print(
            "MISCONCEPTION STATUS:",
            model[
                "misconception_status"
            ]
        )


        print(
            "ACTION:",
            model[
                "next_action"
            ]
        )


        print(
            "DECISION REASON:",
            model[
                "decision_reason"
            ]
        )


        print(
            "EXPECTED NEXT EVIDENCE:",
            model[
                "expected_next_evidence"
            ]
        )


        print(
            "ADVANCE ALLOWED:",
            model[
                "advance_allowed"
            ]
        )


        print(
            "EVIDENCE MATRIX:",
            model[
                "evidence_matrix"
            ]
        )


        print(
            "CHECKS:",
            checks
        )


        print(
            "RESULT:",
            "PASS"
            if passed
            else "FAIL"
        )


    return result


# ================================================================
# RUN T01-T15
# ================================================================

def run_fixed_benchmark(
    show=True
):


    results = []


    for number in range(
        1,
        16
    ):


        test_id = (

            "T"
            + str(number).zfill(2)
        )


        results.append(

            run_validation_case(

                test_id,

                show=show
            )
        )


    return results

def classify_for_adversarial(

    question,

    response,

    expected_types,

    expected_text,

    history=""
):


    raw = classify_student_response(

        current_concept,

        learning_objective,

        question,

        response,

        history,

        expected_types,

        expected_text
    )


    return parse_classification(
        raw
    )

def test_order_sensitivity():


    model_a = create_student_model()


    model_b = create_student_model()


    events_a = [

        (
            "CORRECT",

            "PROBE",

            "DIAGNOSIS",

            [
                "PREDICTION"
            ]
        ),

        (
            "CORRECT",

            "PROBE",

            "DIAGNOSIS",

            [
                "EXPLANATION"
            ]
        ),

        (
            "CORRECT",

            "RETEST",

            "RETEST",

            [
                "INDEPENDENT_RETEST",

                "EXPLANATION"
            ]
        ),

        (
            "CORRECT",

            "TRANSFER",

            "TRANSFER",

            [
                "TRANSFER",

                "EXPLANATION"
            ]
        )
    ]


    events_b = [

        events_a[1],

        events_a[0],

        events_a[2],

        events_a[3]
    ]


    for (

        classification,

        action,

        phase,

        expected_types

    ) in events_a:


        model_a = update_evidence_matrix(

            model_a,

            classification,

            action,

            phase,

            expected_types
        )


    for (

        classification,

        action,

        phase,

        expected_types

    ) in events_b:


        model_b = update_evidence_matrix(

            model_b,

            classification,

            action,

            phase,

            expected_types
        )


    return {

        "id":
            "A01",

        "name":
            "Order sensitivity",

        "passed":

            model_a[
                "evidence_matrix"
            ]

            ==

            model_b[
                "evidence_matrix"
            ]
    }


def test_wording_variation():


    (

        classification,

        _,

        misconception_id,

        _

    ) = classify_for_adversarial(


        (
            "Are north and south poles "
            "the same as positive "
            "and negative electric charges?"
        ),


        (
            "The north end is basically the plus side "
            "and the south end is basically the minus side."
        ),


        [
            "EXPLANATION"
        ],


        (
            "The learner separates magnetic pole names "
            "from electric charge."
        )
    )


    return {

        "id":
            "A02",

        "name":
            "Wording variation",

        "passed":

            classification
            == "MISCONCEPTION"

            and

            misconception_id
            == "MAGNETIC_POLES_ARE_ELECTRIC_CHARGES"
    }


def test_short_answer():


    (

        classification,

        _,

        _,

        _

    ) = classify_for_adversarial(


        (
            "An unknown pole attracts a north pole. "
            "Which pole is it, and why?"
        ),


        "Attract.",


        [
            "EXPLANATION"
        ],


        (
            "The learner identifies the pole "
            "and explains why."
        )
    )


    return {

        "id":
            "A03",

        "name":
            "Short answer",

        "passed":

            classification
            in [

                "INSUFFICIENT_EVIDENCE",

                "PARTIALLY_CORRECT"
            ]
    }


# ================================================================
# A04 - LUCKY GUESS
# ================================================================

def test_lucky_guess():


    (

        classification,

        _,

        _,

        _

    ) = classify_for_adversarial(


        (
            "Two magnets repel. "
            "A is North. "
            "What is B, and why?"
        ),


        "North.",


        [
            "TRANSFER",

            "EXPLANATION"
        ],


        (
            "The learner identifies B "
            "and explains "
            "the changed-context inference."
        )
    )


    return {

        "id":
            "A04",

        "name":
            "Lucky guess",

        "passed":

            classification
            in [

                "INSUFFICIENT_EVIDENCE",

                "PARTIALLY_CORRECT"
            ]
    }


# ================================================================
# A05 - HINT DEPENDENCE
# ================================================================

def test_hint_dependence():


    result = run_validation_case(

        "T07",

        show=False
    )


    model = result[
        "model"
    ]


    return {

        "id":
            "A05",

        "name":
            "Hint dependence",

        "passed":

            model[
                "next_action"
            ]
            == "RETEST"

            and

            not model[
                "advance_allowed"
            ]
    }


# ================================================================
# A06 - RETEST FAILURE
# ================================================================

def test_retest_failure():


    result = run_validation_case(

        "T08",

        show=False
    )


    model = result[
        "model"
    ]


    return {

        "id":
            "A06",

        "name":
            "Retest failure",

        "passed":

            model[
                "evidence_matrix"
            ][
                "independent_retest"
            ]
            == "FAILED"

            and

            not model[
                "advance_allowed"
            ]
    }


# ================================================================
# A07 - TRANSFER FAILURE
# ================================================================

def test_transfer_failure():


    result = run_validation_case(

        "T10",

        show=False
    )


    model = result[
        "model"
    ]


    return {

        "id":
            "A07",

        "name":
            "Transfer failure",

        "passed":

            model[
                "evidence_matrix"
            ][
                "transfer"
            ]
            == "FAILED"

            and

            not model[
                "advance_allowed"
            ]
    }


# ================================================================
# A08 - CONTRADICTION AWARENESS
# ================================================================

def test_contradiction_awareness():


    model = create_student_model()


    model[
        "student_state"
    ] = "DEVELOPING"


    model[
        "phase"
    ] = "DIAGNOSIS"


    model[
        "next_action"
    ] = "PROBE"


    set_expected(

        model,

        [
            "EXPLANATION"
        ],

        (
            "The learner explains "
            "the magnetic-pole rule."
        )
    )


    history = (

        "Student:\n"

        "North and north repel "
        "because like poles repel.\n\n"

        "Tutor:\n"

        "What made you choose that?"
    )


    classification_result = (
        classify_student_response(

            current_concept,

            learning_objective,

            (
                "What happens when two north poles "
                "face each other, and why?"
            ),

            (
                "North and north attract "
                "because both are magnets."
            ),

            history,

            model[
                "expected_evidence_types"
            ],

            model[
                "expected_next_evidence"
            ]
        )
    )


    model = update_tutor_state(

        classification_result,

        model
    )


    reply = (
        generate_tutor_response(

            model,

            classification_result,

            (
                "North and north attract "
                "because both are magnets."
            ),

            history
        )
        .lower()
    )


    markers = [

        "earlier",

        "before",

        "you said",

        "previous",

        "now you're saying",

        "now you are saying"
    ]


    return {

        "id":
            "A08",

        "name":
            "Contradiction awareness",

        "passed":

            any(

                marker
                in reply

                for marker
                in markers
            )
    }


# ================================================================
# A09 - SCIENTIFIC WORDING
# ================================================================

def test_scientific_wording():


    model = prepare_validation_state(
        "T13"
    )


    case = VALIDATION_CASES[
        "T13"
    ]


    classification_result = (
        classify_student_response(

            current_concept,

            learning_objective,

            case[
                "question"
            ],

            case[
                "response"
            ],

            "",

            model[
                "expected_evidence_types"
            ],

            model[
                "expected_next_evidence"
            ]
        )
    )


    model = update_tutor_state(

        classification_result,

        model
    )


    reply = (
        generate_tutor_response(

            model,

            classification_result,

            case[
                "response"
            ],

            ""
        )
        .lower()
    )


    useful_clarification = (

        "represent"
        in reply

        or

        "way scientists show"
        in reply

        or

        "not physical"
        in reply
    )


    return {

        "id":
            "A09",

        "name":
            "Scientific wording",

        "passed":

            model[
                "last_classification"
            ]
            == "MISCONCEPTION"

            and

            useful_clarification
    }


# ================================================================
# A10 - REPEATABILITY
# ================================================================

def test_repeatability():


    critical = [

        "T02",

        "T03",

        "T05",

        "T10",

        "T12",

        "T13"
    ]


    stable = True


    details = {}


    for test_id in critical:


        signatures = []


        for _ in range(
            REPEATABILITY_RUNS
        ):


            result = (
                run_validation_case(

                    test_id,

                    show=False
                )
            )


            model = result[
                "model"
            ]


            signatures.append(

                (

                    model[
                        "last_classification"
                    ],

                    model[
                        "last_semantic_event"
                    ],

                    model[
                        "student_state"
                    ],

                    model[
                        "next_action"
                    ],

                    model[
                        "misconception_id"
                    ],

                    model[
                        "misconception_status"
                    ],

                    model[
                        "advance_allowed"
                    ]
                )
            )


        case_stable = all(

            signature
            == signatures[0]

            for signature
            in signatures
        )


        details[
            test_id
        ] = case_stable


        if not case_stable:


            stable = False


    return {

        "id":
            "A10",

        "name":
            "Repeatability",

        "passed":
            stable,

        "details":
            details
    }


# ================================================================
# RUN ADVERSARIAL SUITE
# ================================================================

def run_adversarial_suite(
    show=True
):


    results = [

        test_order_sensitivity(),

        test_wording_variation(),

        test_short_answer(),

        test_lucky_guess(),

        test_hint_dependence(),

        test_retest_failure(),

        test_transfer_failure(),

        test_contradiction_awareness(),

        test_scientific_wording(),

        test_repeatability()
    ]


    if show:


        print(
            "\n"
            + "=" * 72
        )


        print(
            "ADVERSARIAL RELIABILITY"
        )


        print(
            "=" * 72
        )


        for result in results:


            print(

                result[
                    "id"
                ],

                result[
                    "name"
                ],

                ":",

                "PASS"
                if result[
                    "passed"
                ]

                else

                "FAIL"
            )


            if (
                "details"
                in result
            ):


                print(
                    "   ",
                    result[
                        "details"
                    ]
                )


    return results


# ================================================================
# FIND RESULT
# ================================================================

def find_result(

    results,

    test_id
):


    for result in results:


        if result[
            "id"
        ] == test_id:


            return result


    return None


# ================================================================
# ACCEPTANCE CRITERIA
# ================================================================

def calculate_acceptance_criteria(

    fixed_results,

    adversarial_results
):


    t01 = find_result(
        fixed_results,
        "T01"
    )


    t05 = find_result(
        fixed_results,
        "T05"
    )


    t07 = find_result(
        fixed_results,
        "T07"
    )


    t08 = find_result(
        fixed_results,
        "T08"
    )


    t09 = find_result(
        fixed_results,
        "T09"
    )


    t11 = find_result(
        fixed_results,
        "T11"
    )


    t15 = find_result(
        fixed_results,
        "T15"
    )


    repeatability = find_result(
        adversarial_results,
        "A10"
    )


    return {


        "AC1 All 15 cases match":

            all(

                result[
                    "checks"
                ][
                    "classification"
                ]

                for result
                in fixed_results
            ),


        "AC2 Incomplete answer is not mastery":

            (

                t05[
                    "model"
                ][
                    "last_classification"
                ]
                != "CORRECT"

                and

                not t05[
                    "model"
                ][
                    "advance_allowed"
                ]
            ),


        "AC3 One correct answer is not mastery":

            (

                not t01[
                    "model"
                ][
                    "advance_allowed"
                ]

                and

                t07[
                    "model"
                ][
                    "misconception_status"
                ]
                != "RESOLVED"
            ),


        "AC4 Expected evidence controls action":

            all(

                result[
                    "checks"
                ][
                    "action"
                ]

                and

                result[
                    "checks"
                ][
                    "expected_evidence_control"
                ]

                for result
                in fixed_results
            ),


        "AC5 Missing evidence blocks advance":

            all(

                result[
                    "checks"
                ][
                    "advance"
                ]

                for result
                in fixed_results
            ),


        "AC6 Correct after hint triggers retest":

            (

                t07[
                    "model"
                ][
                    "next_action"
                ]
                == "RETEST"

                and

                t07[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "independent_retest"
                ]
                == "PENDING"
            ),


        "AC7 Retest failure blocks advance":

            (

                t08[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "independent_retest"
                ]
                == "FAILED"

                and

                not t08[
                    "model"
                ][
                    "advance_allowed"
                ]
            ),


        "AC8 Genuine changed-context transfer":

            (

                t09[
                    "model"
                ][
                    "next_action"
                ]
                == "TRANSFER"

                and

                t11[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "transfer"
                ]
                == "PASSED"

                and

                t15[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "transfer"
                ]
                == "PASSED"
            ),


        "AC9 No ACTIVE directly to RESOLVED":

            (

                t07[
                    "model"
                ][
                    "misconception_status"
                ]
                in [

                    "ACTIVE",

                    "REDUCED"
                ]

                and

                t11[
                    "model"
                ][
                    "misconception_status"
                ]
                == "RESOLVED"
            ),


        "AC10 Support and phase are separate":

            all(

                1
                <= result[
                    "model"
                ][
                    "support_level"
                ]
                <= 6

                and

                result[
                    "model"
                ][
                    "phase"
                ]
                in [

                    "DIAGNOSIS",

                    "GUIDED_PRACTICE",

                    "RETEST",

                    "TRANSFER",

                    "MASTERY_CHECK"
                ]

                for result
                in fixed_results
            ),


        "AC11 Decision reason accurately explains action":

            all(

                result[
                    "checks"
                ][
                    "decision_reason"
                ]

                for result
                in fixed_results
            ),


        "AC12 Repeatability":

            repeatability[
                "passed"
            ],


        "AC13 Field-line scientific accuracy":

            find_result(

                adversarial_results,

                "A09"
            )[
                "passed"
            ]
    }


# ================================================================
# G1-G10
# ================================================================

def calculate_gates(

    fixed_results,

    adversarial_results,

    acceptance
):


    t02 = find_result(
        fixed_results,
        "T02"
    )


    t07 = find_result(
        fixed_results,
        "T07"
    )


    t08 = find_result(
        fixed_results,
        "T08"
    )


    t09 = find_result(
        fixed_results,
        "T09"
    )


    t10 = find_result(
        fixed_results,
        "T10"
    )


    t11 = find_result(
        fixed_results,
        "T11"
    )


    t12 = find_result(
        fixed_results,
        "T12"
    )


    t13 = find_result(
        fixed_results,
        "T13"
    )


    t14 = find_result(
        fixed_results,
        "T14"
    )


    t15 = find_result(
        fixed_results,
        "T15"
    )


    return {


        "G1 Classification":

            acceptance[
                "AC1 All 15 cases match"
            ],


        "G2 Misconception lifecycle":

            (

                t02[
                    "model"
                ][
                    "misconception_status"
                ]
                == "ACTIVE"

                and

                t07[
                    "model"
                ][
                    "misconception_status"
                ]
                == "REDUCED"

                and

                t11[
                    "model"
                ][
                    "misconception_status"
                ]
                == "RESOLVED"

                and

                t14[
                    "model"
                ][
                    "misconception_status"
                ]
                != "RESOLVED"
            ),


        "G3 Evidence accumulation":

            (

                t08[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "independent_retest"
                ]
                == "FAILED"

                and

                t09[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "independent_retest"
                ]
                == "PASSED"

                and

                t10[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "transfer"
                ]
                == "FAILED"

                and

                t11[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "transfer"
                ]
                == "PASSED"

                and

                t15[
                    "model"
                ][
                    "evidence_matrix"
                ][
                    "transfer"
                ]
                == "PASSED"
            ),


        "G4 Expected Evidence gating":

            acceptance[
                "AC4 Expected evidence controls action"
            ],


        "G5 Support vs Phase separation":

            acceptance[
                "AC10 Support and phase are separate"
            ],


        "G6 Independent Retest":

            (

                acceptance[
                    "AC6 Correct after hint triggers retest"
                ]

                and

                acceptance[
                    "AC7 Retest failure blocks advance"
                ]
            ),


        "G7 Transfer":

            acceptance[
                "AC8 Genuine changed-context transfer"
            ],


        "G8 Advancement control":

            acceptance[
                "AC5 Missing evidence blocks advance"
            ],


        "G9 Adversarial reliability":

            all(

                result[
                    "passed"
                ]

                for result
                in adversarial_results
            ),


        "G10 Scientific accuracy":

            (

                t12[
                    "model"
                ][
                    "last_classification"
                ]
                == "MISCONCEPTION"

                and

                t13[
                    "model"
                ][
                    "last_classification"
                ]
                == "MISCONCEPTION"

                and

                acceptance[
                    "AC13 Field-line scientific accuracy"
                ]
            )
    }


# ================================================================
# FINAL REPORT
# ================================================================

def print_final_validation_report(

    fixed_results,

    adversarial_results,

    acceptance,

    gates
):


    print(
        "\n\n"
        + "=" * 76
    )


    print(
        "V3.1 FINAL VALIDATION REPORT"
    )


    print(
        "=" * 76
    )


    print(
        "\nT01-T15:"
    )


    for result in fixed_results:


        print(

            result[
                "id"
            ],

            ":",

            "PASS"
            if result[
                "passed"
            ]

            else

            "FAIL"
        )


    print(
        "\nADVERSARIAL:"
    )


    for result in adversarial_results:


        print(

            result[
                "id"
            ],

            result[
                "name"
            ],

            ":",

            "PASS"
            if result[
                "passed"
            ]

            else

            "FAIL"
        )


    print(
        "\nACCEPTANCE CRITERIA:"
    )


    for name, status in (
        acceptance.items()
    ):


        print(

            name,

            ":",

            "PASS"
            if status
            else
            "FAIL"
        )


    print(
        "\nG1-G10:"
    )


    for name, status in (
        gates.items()
    ):


        print(

            name,

            ":",

            "PASS"
            if status
            else
            "FAIL"
        )


    final_pass = (

        all(

            result[
                "passed"
            ]

            for result
            in fixed_results
        )

        and

        all(

            result[
                "passed"
            ]

            for result
            in adversarial_results
        )

        and

        all(
            acceptance.values()
        )

        and

        all(
            gates.values()
        )
    )


    print(
        "\n"
        + "=" * 76
    )


    if final_pass:


        print(
            "V3.1 VALIDATION GATE: PASS"
        )


        print(
            "V3.1 IS ELIGIBLE FOR FREEZE."
        )


    else:


        print(
            "V3.1 VALIDATION GATE: FAIL"
        )


        print(
            "DO NOT START V3.2."
        )


        print(
            "Fix only the failed decision logic, "
            "then rerun the same gate."
        )


    print(
        "=" * 76
    )


    return final_pass


def run_complete_v31_validation():


    fixed_results = (
        run_fixed_benchmark(
            show=True
        )
    )


    adversarial_results = (
        run_adversarial_suite(
            show=True
        )
    )


    acceptance = (
        calculate_acceptance_criteria(

            fixed_results,

            adversarial_results
        )
    )


    gates = (
        calculate_gates(

            fixed_results,

            adversarial_results,

            acceptance
        )
    )


    passed = (
        print_final_validation_report(

            fixed_results,

            adversarial_results,

            acceptance,

            gates
        )
    )


    return {

        "fixed_results":
            fixed_results,

        "adversarial_results":
            adversarial_results,

        "acceptance":
            acceptance,

        "gates":
            gates,

        "passed":
            passed
    }



if __name__ == "__main__":


    if MODE == "VALIDATION":


        validation_results = (
            run_complete_v31_validation()
        )


    else:


        run_tutor()
```
